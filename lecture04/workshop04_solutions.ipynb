## Toolbox

The functions you need today. The lecture notebook shows all of them in use.

## Code patterns

Four complete patterns for today, one for each test. Copy a pattern into a task cell, replace `df`, the column names
and the group names, and choose the p-value line that matches your H<sub>1</sub>. The patterns are templates; they do
not run as they stand.

**Paired t test (the same units measured twice)**

```python
d = df["after"] - df["before"]               # one difference per row
n = len(d)
t_stat = d.mean() / (d.std(ddof=1) / np.sqrt(n))
p = stats.t.sf(t_stat, df=n - 1)             # H1: mu_d > 0
# H1: mu_d < 0   p = stats.t.cdf(t_stat, df=n - 1)
# H1: mu_d != 0  p = 2 * stats.t.sf(abs(t_stat), df=n - 1), critical values -/+ stats.t.ppf(1 - alpha / 2, df=n - 1)
print(f"d_bar = {d.mean():.3f}, t = {t_stat:.3f}, p = {p:.4f}")
```

**t test for two independent means (Welch)**

```python
x1 = df.loc[df["group"] == "first", "column"]
x2 = df.loc[df["group"] == "second", "column"]
n1, n2 = len(x1), len(x2)
v1 = x1.var(ddof=1) / n1
v2 = x2.var(ddof=1) / n2
se = np.sqrt(v1 + v2)                        # variances add
t_stat = (x1.mean() - x2.mean()) / se
df_w = (v1 + v2)**2 / (v1**2 / (n1 - 1) + v2**2 / (n2 - 1))
p = stats.t.sf(t_stat, df=df_w)              # H1: mu1 - mu2 > 0; two-sided: 2 * stats.t.sf(abs(t_stat), df=df_w)
print(f"t = {t_stat:.3f}, df = {df_w:.2f}, p = {p:.4f}")
```

**z test for two proportions (pooled share, as in Newbold)**

```python
x1 = df.loc[df["group"] == "first", "column"]     # a 0/1 column
x2 = df.loc[df["group"] == "second", "column"]
n1, n2 = len(x1), len(x2)
p0 = (x1.sum() + x2.sum()) / (n1 + n2)            # pooled share
print(n1 * p0 * (1 - p0), n2 * p0 * (1 - p0))     # the condition: both above 5
z = (x1.mean() - x2.mean()) / np.sqrt(p0 * (1 - p0) * (1 / n1 + 1 / n2))
p = stats.norm.sf(z)                              # H1: P1 > P2; two-sided: 2 * stats.norm.sf(abs(z))
print(f"z = {z:.3f}, p = {p:.4f}")
```

**F test for two variances (needs normal populations)**

```python
F = x1.var(ddof=1) / x2.var(ddof=1)          # put the larger variance on top
nu1, nu2 = len(x1) - 1, len(x2) - 1          # numerator first
crit = stats.f.ppf(1 - alpha, nu1, nu2)      # one-sided; two-sided: stats.f.ppf(1 - alpha / 2, nu1, nu2)
p = stats.f.sf(F, nu1, nu2)                  # one-sided; two-sided: 2 * stats.f.sf(F, nu1, nu2)
print(f"F = {F:.3f}, crit = {crit:.3f}, p = {p:.4f}")
```

### Task 1. Copenhagen against Aarhus in code (worked)

Do orders shipped from Copenhagen take longer on average than orders shipped from Aarhus? The orders in the two groups
are different orders, so the samples are independent. The hypotheses are H<sub>0</sub>: μ<sub>1</sub> − μ<sub>2</sub>
≤ 0 and H<sub>1</sub>: μ<sub>1</sub> − μ<sub>2</sub> > 0, with group 1 Copenhagen and group 2 Aarhus. The cell below
runs Welch's test exactly as in the lecture. Run it and answer the questions.

1. Which lines compute the standard error? Why are the two parts added, and why is the result larger than either
group's own standard error?
2. Is Carrier B slower than Carrier A? Change the first two lines to `column = "carrier"` and
`group1, group2 = "Carrier B", "Carrier A"` and run the cell again. What do you conclude?
3. Write the result for the two warehouses in one or two sentences for a manager at RetailCorp. Include what the
comparison cannot show.

````python
column = "warehouse"                          # question 2: "carrier"
group1, group2 = "Copenhagen", "Aarhus"       # question 2: "Carrier B", "Carrier A"

x1 = sample.loc[sample[column] == group1, "delivery_hours"]
x2 = sample.loc[sample[column] == group2, "delivery_hours"]
n1, n2 = len(x1), len(x2)
print(f"{group1}: n = {n1}, mean = {x1.mean():.3f}, s = {x1.std(ddof=1):.3f}")
print(f"{group2}: n = {n2}, mean = {x2.mean():.3f}, s = {x2.std(ddof=1):.3f}")

v1 = x1.var(ddof=1) / n1                      # s1^2 / n1
v2 = x2.var(ddof=1) / n2                      # s2^2 / n2
se = np.sqrt(v1 + v2)                         # variances add
t_stat = (x1.mean() - x2.mean()) / se
df_w = (v1 + v2)**2 / (v1**2 / (n1 - 1) + v2**2 / (n2 - 1))   # Welch degrees of freedom
t_crit = stats.t.ppf(1 - alpha, df=df_w)
p = stats.t.sf(t_stat, df=df_w)               # area above t, since H1 has >
print(f"se = {se:.4f}, t = {t_stat:.3f}, df = {df_w:.2f}, crit = {t_crit:.3f}, p = {p:.3f}")
````

**Answer.**
1. The lines with `v1`, `v2` and `se`. The standard error is √(0.8889 + 0.5116) = 1.1834 hours. Each sample average
has its own chance error, and the difference carries both. Because the samples are independent the errors do not
cancel, and the variances add. So 1.1834 is larger than the standard error of either average (0.943 and 0.715 hours),
but smaller than their sum.
2. Carrier B: mean 37.208 hours (73 orders); Carrier A: 36.548 hours (127 orders). t = 0.577 with 171.33 degrees of
freedom and p = 0.282, so we do not reject H<sub>0</sub>. The sample does not give strong evidence that Carrier B is
slower on average. It does not show that the carriers are equally fast either.
3. In the sample, Copenhagen orders took 2.3 hours longer on average (38.2 against 35.9 hours). At the 5% level this
is evidence that Copenhagen orders take longer on average (p = 0.025). The orders were not assigned to the warehouses
at random, so the test shows that the warehouses differ. It cannot say why; other differences between the orders
could play a part.

For the TAs: if students run `stats.ttest_ind(x1, x2, equal_var=False)` without `alternative`, they get the
two-sided p-value, 0.04999..., printed as 0.050. The question is one-sided, so p = 0.025 is the right one. A
two-sided test would sit exactly at the 5% line here, which is a good moment to say why the direction is decided from
the question before looking at the data.

### Task 2. Are more Copenhagen orders late?

An order is late if it takes more than 48 hours (column `late`, 1 for late). Operations suspects that a larger share
of Copenhagen orders is late than of Aarhus orders. Test this at α = 0.05 with the 200 orders.

**Answer in words.** Are the samples paired or independent? State H<sub>0</sub> and H<sub>1</sub>, check the
condition, and give the decision and the result for operations.

<details>
<summary>Hint 1</summary>

This is the z test for two proportions with the pooled share p̂<sub>0</sub>, one-sided: H<sub>0</sub>:
P<sub>Cph</sub> − P<sub>Aar</sub> ≤ 0 and H<sub>1</sub>: P<sub>Cph</sub> − P<sub>Aar</sub> > 0. The orders in the two
groups are different orders, so the samples are independent.
</details>

<details>
<summary>Hint 2</summary>

`sample.loc[sample["warehouse"] == "Copenhagen", "late"]`, `.sum()`, `.mean()`, `len`, `np.sqrt`,
`stats.norm.ppf(1 - alpha)` and `stats.norm.sf`. See the third code pattern.
</details>

<details>
<summary>Hint 3</summary>

```python
# select the late column for Copenhagen and for Aarhus
# n1, n2 and the two sample shares
# the pooled share: all late orders divided by all orders
# the condition n * p0 * (1 - p0) for each group
# the standard error with p0, then z
# the critical value and the p-value (area above z)
```
</details>

````python
x1 = sample.loc[sample["warehouse"] == "Copenhagen", "late"]
x2 = sample.loc[sample["warehouse"] == "Aarhus", "late"]
n1, n2 = len(x1), len(x2)
p0 = (x1.sum() + x2.sum()) / (n1 + n2)
print(f"Copenhagen: {x1.sum()} of {n1} late ({x1.mean():.3f}); Aarhus: {x2.sum()} of {n2} late ({x2.mean():.3f})")
print(f"Pooled share {p0:.3f}, condition {n1 * p0 * (1 - p0):.1f} and {n2 * p0 * (1 - p0):.1f}")
se = np.sqrt(p0 * (1 - p0) * (1 / n1 + 1 / n2))
z = (x1.mean() - x2.mean()) / se
print(f"se = {se:.5f}, z = {z:.3f}, crit = {stats.norm.ppf(1 - alpha):.3f}, p = {stats.norm.sf(z):.3f}")
````

**Answer.** The samples are independent: different orders in each group. H<sub>0</sub>: P<sub>Cph</sub> −
P<sub>Aar</sub> ≤ 0 and H<sub>1</sub>: P<sub>Cph</sub> − P<sub>Aar</sub> > 0, one-sided because operations suspects a
*larger* share in Copenhagen.

10 of 80 Copenhagen orders were late (0.125) against 6 of 120 Aarhus orders (0.050). The pooled share is
16/200 = 0.080. The condition gives 80 × 0.08 × 0.92 = 5.9 and 120 × 0.08 × 0.92 = 8.8, both above 5, though the
Copenhagen value is close to the limit, so the normal approximation is only just acceptable. The standard error is
0.03916, z = 1.915 is above 1.645, and p = 0.028, so we reject H<sub>0</sub>.

For operations: in the sample, 12.5% of Copenhagen orders were late against 5% of Aarhus orders. At the 5% level this
is evidence that a larger share of Copenhagen orders is late. The p-value is not far from the 5% line and rests on 16
late orders in total, so a larger study would give a firmer answer. The comparison does not say why Copenhagen
differs.

Each passenger gave both ratings, so the samples are paired. Compute the difference for each passenger and use the

<details>

# d = check-in rating minus baggage rating, one value per passenger

For the service manager: if the file can be treated as a random sample of the airline's passengers, passengers rate

business question. The ratings are scores on a short scale that we treat as numbers, and the source does not say how

### Task 4. Do the two carriers differ in spread?

A larger spread in delivery times makes deliveries less predictable. Do the delivery times of Carrier A and Carrier B
differ in spread, in either direction? Test at α = 0.05. The test assumes normal populations, so look at the two
histograms first.

**Answer in words.** State H<sub>0</sub> and H<sub>1</sub>. Is the normality assumption reasonable? Which variance
did you put on top, and why? Give the decision and what it tells RetailCorp, and what it does not tell.

<details>
<summary>Hint 1</summary>

The F test for two variances, two-sided: H<sub>0</sub>: σ<sub>A</sub>² = σ<sub>B</sub>² and H<sub>1</sub>:
σ<sub>A</sub>² ≠ σ<sub>B</sub>². As in Newbold, put the larger sample variance on top, compare F with the critical
value at α/2, and double the area above F for the p-value.
</details>

<details>
<summary>Hint 2</summary>

`sample.loc[sample["carrier"] == "Carrier A", "delivery_hours"]`, `plt.hist`, `stats.skew`, `.var(ddof=1)`,
`stats.f.ppf(1 - alpha / 2, nu1, nu2)` and `2 * stats.f.sf(F, nu1, nu2)`. See the fourth code pattern.
</details>

<details>
<summary>Hint 3</summary>

```python
# select the delivery times for each carrier
# histograms and skewness for both
# the two sample variances: which is larger?
# F = larger variance / smaller variance, with nu1 and nu2 in the same order
# the critical value at alpha / 2 and the two-sided p-value
```
</details>

````python
a = sample.loc[sample["carrier"] == "Carrier A", "delivery_hours"]
b = sample.loc[sample["carrier"] == "Carrier B", "delivery_hours"]
fig, axes = plt.subplots(1, 2, figsize=(10, 3), sharex=True)
for ax, x, name in [(axes[0], a, "Carrier A"), (axes[1], b, "Carrier B")]:
    ax.hist(x, bins=15, edgecolor="white")
    ax.set_title(f"{name}: skewness {stats.skew(x):.2f}")
plt.show()

print(f"Carrier A: n = {len(a)}, s^2 = {a.var(ddof=1):.3f}; Carrier B: n = {len(b)}, s^2 = {b.var(ddof=1):.3f}")
F = a.var(ddof=1) / b.var(ddof=1)            # Carrier A has the larger variance
nu1, nu2 = len(a) - 1, len(b) - 1
crit = stats.f.ppf(1 - alpha / 2, nu1, nu2)
p = 2 * stats.f.sf(F, nu1, nu2)
print(f"F = {F:.3f}, crit = {crit:.3f}, p (two-sided) = {p:.3f}")
````

**Answer.** H<sub>0</sub>: σ<sub>A</sub>² = σ<sub>B</sub>² and H<sub>1</sub>: σ<sub>A</sub>² ≠ σ<sub>B</sub>². Both
histograms are roughly symmetric and bell-shaped (skewness −0.08 and 0.25), so the normality assumption is reasonable.

Carrier A has the larger sample variance (74.257 against 52.895), so it goes on top: F = 1.404 with 126 and 72
degrees of freedom. The critical value F<sub>126, 72, 0.025</sub> = 1.528 and the two-sided p = 0.116, so we do not
reject H<sub>0</sub>.

For RetailCorp: the sample does not give strong evidence that the carriers differ in how spread out their delivery
times are. This does not show that they are equal.

For the TAs: in the population file the two carriers have practically the same standard deviation (7.78 and 7.80
hours), so here the non-rejection is the correct decision. Yet the sample ratio of 1.40 is larger than the warehouse
ratio in the lecture (1.16), where the population variances do differ. Sample variances vary a lot at these sample
sizes, which is why the F test has little power here.

### Extension 1. How much does pairing help?

In Task 3 you used the paired test. Analyze the same two airline columns as if they were two independent groups, with
the standard error √(s<sub>1</sub>²/n + s<sub>2</sub>²/n), and compare the standard error and t with Task 3. Then
compute the correlation between the two ratings with `airline["Checkin service"].corr(airline["Baggage handling"])`.

**Answer in words.** Why does pairing change the standard error much less here than in the RetailCorp survey in the
lecture, where the correlation between before and after was 0.94?

<details>
<summary>Hint 1</summary>

The independent standard error uses the variance of each column; the paired one uses the variance of the
differences. The variance of a difference is smaller when the two measurements move together.
</details>

<details>
<summary>Hint 2</summary>

`x1.var(ddof=1)`, `x2.var(ddof=1)`, `np.sqrt`, `.corr()`.
</details>

````python
x1, x2 = airline["Checkin service"], airline["Baggage handling"]
n = len(x1)
se_ind = np.sqrt(x1.var(ddof=1) / n + x2.var(ddof=1) / n)
se_pair = (x1 - x2).std(ddof=1) / np.sqrt(n)
print(f"Independent: se = {se_ind:.5f}, t = {(x1.mean() - x2.mean()) / se_ind:.2f}")
print(f"Paired:      se = {se_pair:.5f}, t = {(x1 - x2).mean() / se_pair:.2f}")
print(f"Correlation: {x1.corr(x2):.2f}")
````

**Answer.** As two independent groups the standard error is 0.01074 and t = −29.71; paired it is 0.00937 and
t = −34.07. Pairing helps a little. The correlation between the two ratings is only 0.24: a passenger who rates
check-in high does not tend much to rate baggage handling high. Pairing removes the part of the variation that the two
measurements share, and here they share little. In the RetailCorp survey a customer's before and after scores were
strongly correlated (0.94), so pairing removed most of the variation and the standard error fell from 0.200 to
0.051. Even when pairing helps little, the paired test is the right one when the data are paired.

### Extension 2. Large samples in the airline file

1. Test whether the share of satisfied passengers (column `satisfaction` equal to "satisfied") differs between women
and men (column `Gender`), two-sided.
2. Test whether average seat comfort differs between women and men, two-sided, with Welch's test.

**Answer in words.** Compare the two results. What does a sample of 26,000 passengers do, and what does it not do?

<details>
<summary>Hint 1</summary>

Part 1 is a two-proportion z test with the pooled share; part 2 is Welch's t test. Both are two-sided.
</details>

<details>
<summary>Hint 2</summary>

`(airline["satisfaction"] == "satisfied")` gives a 0/1 column. `stats.ttest_ind(women, men, equal_var=False)`.
</details>

````python
sat = airline["satisfaction"] == "satisfied"
w, m_ = sat[airline["Gender"] == "Female"], sat[airline["Gender"] == "Male"]
p0 = (w.sum() + m_.sum()) / (len(w) + len(m_))
z = (w.mean() - m_.mean()) / np.sqrt(p0 * (1 - p0) * (1 / len(w) + 1 / len(m_)))
print(f"Satisfied: women {w.mean():.3f}, men {m_.mean():.3f}, z = {z:.2f}, p = {2 * stats.norm.sf(abs(z)):.3f}")

sw = airline.loc[airline["Gender"] == "Female", "Seat comfort"]
sm = airline.loc[airline["Gender"] == "Male", "Seat comfort"]
r = stats.ttest_ind(sw, sm, equal_var=False)
print(f"Seat comfort: women {sw.mean():.3f}, men {sm.mean():.3f}, t = {r.statistic:.2f}, p = {r.pvalue:.1e}")
````

**Answer.**
1. 43.5% of the women and 44.3% of the men are satisfied. z = −1.18 and p = 0.24, so we do not reject H<sub>0</sub>,
even with 26,000 passengers.
2. Average seat comfort is 3.51 for women and 3.39 for men, a difference of 0.13 points on a scale from 1 to 5.
t = 7.70 and p is about 1e-14, so we reject H<sub>0</sub>.

A large sample makes the standard error small, so even small differences become statistically significant, as in
part 2. It does not make every test reject. In part 1 the sample difference is 0.8 percentage points and the test does
not reject; with 26,000 passengers that difference is estimated precisely, and it is small. A significant result says
that a difference would be unlikely if there were none in the population. Whether it is large enough to matter is a
separate question, so always report the size of the difference next to the p-value.

### Extension 3. When pooled and Welch disagree

A study compares a small group (n = 10, mean 52, standard deviation 12) with a large group (n = 60, mean 45, standard
deviation 4). Run the two-sided test with `stats.ttest_ind_from_stats(52, 12, 10, 45, 4, 60)`, first as written (the
default), then with `equal_var=False` added.

**Answer in words.** Compare the two results. Which version is the default, why do they differ so much here, and
which one would you report?

<details>
<summary>Hint 1</summary>

The default is the pooled test. Look at which group's variance dominates the pooled variance, and which group has the
larger variance.
</details>

<details>
<summary>Hint 2</summary>

`stats.ttest_ind_from_stats(mean1, std1, nobs1, mean2, std2, nobs2, equal_var=False)`.
</details>

````python
print(stats.ttest_ind_from_stats(52, 12, 10, 45, 4, 60))
print(stats.ttest_ind_from_stats(52, 12, 10, 45, 4, 60, equal_var=False))
````

**Answer.** The default is the pooled test: t = 3.57 and p = 0.0007, so H<sub>0</sub> would be rejected clearly. With
`equal_var=False` (Welch), t = 1.83 and p = 0.0997, so H<sub>0</sub> is not rejected at 5%.

The pooled variance is a weighted average with weights n − 1, so it is dominated by the large group with the small
spread (s = 4). That understates the uncertainty of the small group's mean, whose spread is three times larger
(s = 12), and the pooled standard error becomes too small. Welch uses each group's own variance. Here the variances
clearly differ, so the pooled assumption does not hold, and Welch is the one to report. In your notebook, say which
version you used and why; with `ttest_ind` the pooled test is what you get if you do not write `equal_var=False`.

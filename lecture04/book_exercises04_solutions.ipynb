# Warm-up (one-sample review)

Before comparing **two** populations, these three warm-ups review the **one-sample** tests from Chapter 9. They are rewritten (new situations and new numbers) from Chapter 9 book exercises that were available for Workshop Lecture 3 but were **not** used in it:

The three statistics are

$$
z = \frac{\bar{x} - \mu_0}{\sigma/\sqrt{n}}, \qquad t = \frac{\bar{x} - \mu_0}{s/\sqrt{n}}\ (df = n-1), \qquad z = \frac{\hat{p} - P_0}{\sqrt{P_0(1-P_0)/n}}
$$

## Warm-up A: Courier delivery times (known $\sigma$, $z$-test)

A courier company claims that same-day parcels are delivered in **at most 4.5 hours** on average. Delivery times are normally distributed with known standard deviation $\sigma = 0.9$ hours. A random sample of $n = 36$ parcels has a mean delivery time of $\bar{x} = 4.82$ hours. Test at $\alpha = 0.05$ whether the mean delivery time exceeds 4.5 hours.

$$
H_0: \mu \le 4.5 \qquad \text{versus} \qquad H_1: \mu > 4.5
$$

```python
import numpy as np
from scipy import stats

mu0 = 4.5
sigma = 0.9
n_A = 36
xbar_A = 4.82
alpha_A = 0.05

se_A = sigma / np.sqrt(n_A)
z_A = (xbar_A - mu0) / se_A
z_crit_A = stats.norm.ppf(1 - alpha_A)
p_A = stats.norm.sf(z_A)   # upper-tailed: P(Z > z)

print(f"SE = σ/√n = {sigma}/√{n_A} = {se_A:.4f}")
print(f"z = ({xbar_A} - {mu0})/{se_A:.4f} = {z_A:.4f}")
print(f"Critical value z_0.05 = {z_crit_A:.4f}  (reject if z > {z_crit_A:.4f})")
print(f"Upper-tailed p-value = {p_A:.4f}")
print("Reject H0" if z_A > z_crit_A else "Do not reject H0")
```

**Conclusion (Warm-up A).** $SE = 0.9/\sqrt{36} = 0.15$, so $z = (4.82 - 4.5)/0.15 = 2.13 > z_{0.05} = 1.645$ and $p \approx 0.016 < 0.05$. We **reject $H_0$**: at the 5% level there is evidence that the mean same-day delivery time exceeds the claimed 4.5 hours.

## Warm-up B: Usefulness of career mentoring ($\sigma$ unknown, $t$-test)

A university careers office asks a random sample of $n = 48$ recent graduates to rate, on a scale from 1 (not useful) to 5 (very useful), how useful career mentoring was. The sample mean rating is $\bar{x} = 3.48$ with sample standard deviation $s = 0.82$. Test at $\alpha = 0.01$ whether the population mean rating is greater than 3.0.

$$
H_0: \mu \le 3.0 \qquad \text{versus} \qquad H_1: \mu > 3.0
$$

```python
mu0_B = 3.0
n_B = 48
xbar_B = 3.48
s_B = 0.82
alpha_B = 0.01

df_B = n_B - 1
se_B = s_B / np.sqrt(n_B)
t_B = (xbar_B - mu0_B) / se_B
t_crit_B = stats.t.ppf(1 - alpha_B, df=df_B)
p_B = stats.t.sf(t_B, df=df_B)   # upper-tailed: P(T > t)

print(f"SE = s/√n = {s_B}/√{n_B} = {se_B:.4f}")
print(f"t = ({xbar_B} - {mu0_B})/{se_B:.4f} = {t_B:.4f},  df = {df_B}")
print(f"Critical value t_(0.01, {df_B}) = {t_crit_B:.4f}  (reject if t > {t_crit_B:.4f})")
print(f"Upper-tailed p-value = {p_B:.6f}")
print("Reject H0" if t_B > t_crit_B else "Do not reject H0")
```

**Conclusion (Warm-up B).** $SE = 0.82/\sqrt{48} \approx 0.1184$, so $t = 0.48/0.1184 \approx 4.06 > t_{47,0.01} \approx 2.408$ and $p \approx 0.0001 < 0.01$. We **reject $H_0$**: even at the 1% level there is strong evidence that graduates' mean usefulness rating of career mentoring is above 3.0.

## Warm-up C: Abandoned online carts and shipping cost (proportion $z$-test)

In a random sample of $n = 280$ online shoppers who abandoned a cart, 98 said that shipping cost was the main reason. Test at $\alpha = 0.05$ whether **more than 30%** of all such shoppers abandon their cart mainly because of shipping cost.

$$
H_0: P \le 0.30 \qquad \text{versus} \qquad H_1: P > 0.30
$$

Under $H_0$ the standard error uses $P_0 = 0.30$: $\;SE = \sqrt{P_0(1-P_0)/n}$. The large-sample condition holds since $nP_0(1-P_0) = 280(0.30)(0.70) = 58.8 > 5$.

```python
n_C = 280
x_C = 98
P0 = 0.30
alpha_C = 0.05

p_hat_C = x_C / n_C
se_C = np.sqrt(P0 * (1 - P0) / n_C)
z_C = (p_hat_C - P0) / se_C
z_crit_C = stats.norm.ppf(1 - alpha_C)
p_C = stats.norm.sf(z_C)   # upper-tailed: P(Z > z)

print(f"p̂ = {x_C}/{n_C} = {p_hat_C:.4f}")
print(f"SE (under H0) = √(P0(1-P0)/n) = {se_C:.4f}")
print(f"z = ({p_hat_C:.4f} - {P0})/{se_C:.4f} = {z_C:.4f}")
print(f"Critical value z_0.05 = {z_crit_C:.4f}  (reject if z > {z_crit_C:.4f})")
print(f"Upper-tailed p-value = {p_C:.4f}")
print("Reject H0" if z_C > z_crit_C else "Do not reject H0")
```

**Conclusion (Warm-up C).** $\hat{p} = 98/280 = 0.35$, $SE = \sqrt{0.30 \cdot 0.70/280} \approx 0.0274$, so $z = 0.05/0.0274 \approx 1.83 > z_{0.05} = 1.645$ and $p \approx 0.034 < 0.05$. We **reject $H_0$**: at the 5% level there is evidence that more than 30% of cart-abandoning shoppers do so mainly because of shipping cost. (At $\alpha = 0.01$ we would not reject, since $p > 0.01$.)

## Part 1 — Paired $t$-test for a mean difference

With **paired** observations we work with the differences $d_i = x_{1i} - x_{2i}$. If the differences are approximately normal and $\sigma_d$ is unknown, the test statistic for $H_0: \mu_1 - \mu_2 = D_0$ is

$$
t = \frac{\bar{d} - D_0}{s_d/\sqrt{n}}, \qquad df = n - 1
$$

where $\bar{d} = \bar{x}_1 - \bar{x}_2$ is the sample mean difference, $s_d$ is the sample standard deviation of the differences and $s_d/\sqrt{n}$ is the standard error.

| $H_1$ | Reject $H_0$ if | $p$-value |
|---|---|---|
| $\mu_1-\mu_2 > 0$ | $t > t_{n-1,\alpha}$ | $P(T > t)$ → `stats.t.sf(t, df)` |
| $\mu_1-\mu_2 < 0$ | $t < -t_{n-1,\alpha}$ | $P(T < t)$ → `stats.t.cdf(t, df)` |
| $\mu_1-\mu_2 \neq 0$ | $|t| > t_{n-1,\alpha/2}$ | $2P(T > |t|)$ |

## Problem

You have been asked to determine if two different production processes have different mean numbers of units produced per hour. Process 1 has mean $\mu_1$ and process 2 has mean $\mu_2$. The hypotheses are:

$$
H_0: \mu_1 - \mu_2 = 0 \qquad \text{versus} \qquad H_1: \mu_1 - \mu_2 > 0
$$

Using a random sample of $n = 25$ paired observations, the sample means are $\bar{x}_1 = 50$ and $\bar{x}_2 = 60$. Can you reject the null hypothesis using a probability of Type I error $\alpha = 0.05$ in each case?

a. The sample standard deviation of the difference is $s_d = 20$.

b. The sample standard deviation of the difference is $s_d = 25$.

```python
import numpy as np
from scipy import stats

n = 25
x1_bar = 50
x2_bar = 60
alpha = 0.05
D0 = 0

d_bar = x1_bar - x2_bar
df = n - 1
t_crit = stats.t.ppf(1 - alpha, df=df)

print(f"Sample mean difference d̄ = {d_bar}")
print(f"Degrees of freedom = {df}")
print(f"Critical value t_(0.05, 24) = {t_crit:.4f}  (upper-tailed: reject if t > {t_crit:.4f})")
```

**Important note before solving.** The alternative is $H_1: \mu_1 - \mu_2 > 0$ (process 1 produces *more*), but the sample gives $\bar{d} = 50 - 60 = -10$, i.e. process 1 produced *less*. The data point in the **opposite direction** from $H_1$, so the $t$-statistic will be negative and the upper-tailed $p$-value will be large. Whatever the value of $s_d$, we **cannot** reject $H_0$ in favour of $H_1: \mu_1-\mu_2>0$.

## Part (a): $s_d = 20$

```python
s_d_a = 20
se_a = s_d_a / np.sqrt(n)
t_a = (d_bar - D0) / se_a
p_a = stats.t.sf(t_a, df=df)   # upper-tailed: P(T > t)

print(f"SE = s_d/√n = {s_d_a}/√{n} = {se_a:.4f}")
print(f"t = {d_bar}/{se_a:.4f} = {t_a:.4f}")
print(f"Critical value = {t_crit:.4f}")
print(f"Upper-tailed p-value = {p_a:.4f}")
print("Reject H0" if t_a > t_crit else "Do not reject H0")
```

**Conclusion (a).** $t = -2.50 < t_{24,0.05} = 1.711$ and $p \approx 0.990 > 0.05$, so we **do not reject $H_0$**. There is no evidence that process 1 has a higher mean output than process 2 — the sample actually shows the opposite.

## Part (b): $s_d = 25$

```python
s_d_b = 25
se_b = s_d_b / np.sqrt(n)
t_b = (d_bar - D0) / se_b
p_b = stats.t.sf(t_b, df=df)   # upper-tailed: P(T > t)

print(f"SE = s_d/√n = {s_d_b}/√{n} = {se_b:.4f}")
print(f"t = {d_bar}/{se_b:.4f} = {t_b:.4f}")
print(f"Critical value = {t_crit:.4f}")
print(f"Upper-tailed p-value = {p_b:.4f}")
print("Reject H0" if t_b > t_crit else "Do not reject H0")
```

**Conclusion (b).** $t = -2.00 < 1.711$ and $p \approx 0.972 > 0.05$, so we **do not reject $H_0$**. A larger $s_d$ makes the standard error larger, so $|t|$ is smaller — but the main reason for not rejecting is again that $\bar{d}$ has the wrong sign for $H_1$.

## Extra: what if the alternative pointed the other way?

The data ($\bar{x}_2 > \bar{x}_1$) suggest process 2 is more productive. Possibly the labels were swapped, or the intended alternative was $H_1: \mu_1-\mu_2<0$ (equivalently $\mu_2-\mu_1>0$) or two-sided. For illustration only:

```python
t_crit_two = stats.t.ppf(1 - alpha/2, df=df)

for label, t_val in [("(a) s_d = 20", t_a), ("(b) s_d = 25", t_b)]:
    p_lower = stats.t.cdf(t_val, df=df)            # H1: mu1 - mu2 < 0
    p_two = 2 * stats.t.sf(abs(t_val), df=df)      # H1: mu1 - mu2 != 0
    print(label)
    print(f"  Lower-tailed: p = {p_lower:.4f} -> {'Reject H0' if t_val < -t_crit else 'Do not reject H0'}  (critical -{t_crit:.4f})")
    print(f"  Two-sided:    p = {p_two:.4f} -> {'Reject H0' if abs(t_val) > t_crit_two else 'Do not reject H0'}  (critical ±{t_crit_two:.4f})")
```

| Case | $t$ | $H_1: \mu_1-\mu_2>0$ (as stated) | $H_1: \mu_1-\mu_2<0$ | $H_1: \mu_1-\mu_2\neq0$ |
|---|---|---|---|---|
| (a) $s_d=20$ | $-2.50$ | $p=0.990$, do not reject | $p=0.010$, reject | $p=0.020$, reject |
| (b) $s_d=25$ | $-2.00$ | $p=0.972$, do not reject | $p=0.028$, reject | $p=0.057$, do not reject |

**Takeaway:** as stated, neither case rejects $H_0$. If the alternative had been $\mu_1-\mu_2<0$, both cases would reject at 5%; two-sided, only (a) would.

# Part 2: Independent samples with known variances ($z$-test)

When the two samples are **independent** and the population variances $\sigma_1^2$ and $\sigma_2^2$ are **known**, the test statistic for $H_0: \mu_1 - \mu_2 = D_0$ is

$$
z = \frac{(\bar{x}_1 - \bar{x}_2) - D_0}{\sqrt{\dfrac{\sigma_1^2}{n_1} + \dfrac{\sigma_2^2}{n_2}}}
$$

For the upper-tailed alternative $H_1: \mu_1 - \mu_2 > 0$, reject $H_0$ if $z > z_\alpha$. The $p$-value is $P(Z > z)$, computed with `norm.sf(z)`.

## Problem

You have been asked to determine if two different production processes have different mean numbers of units produced per hour. Process 1 has mean $\mu_1$ and process 2 has mean $\mu_2$. The hypotheses are:

$$
H_0: \mu_1 - \mu_2 = 0 \qquad \text{versus} \qquad H_1: \mu_1 - \mu_2 > 0
$$

Use a random sample of $n_1 = 25$ observations from process 1 and $n_2 = 28$ observations from process 2. The known variance for process 1 is $\sigma_1^2 = 900$ and the known variance for process 2 is $\sigma_2^2 = 1600$. Can you reject the null hypothesis using a probability of Type I error $\alpha = 0.05$ in each case?

a. The sample means are $\bar{x}_1 = 50$ and $\bar{x}_2 = 60$.

b. The difference in sample means is $\bar{x}_1 - \bar{x}_2 = 20$.

```python
from scipy.stats import norm

n1 = 25
n2 = 28
var1 = 900
var2 = 1600
alpha = 0.05
D0 = 0

se_ind = np.sqrt(var1 / n1 + var2 / n2)
z_crit = norm.ppf(1 - alpha)

print(f"SE = √({var1}/{n1} + {var2}/{n2}) = √({var1/n1:.4f} + {var2/n2:.4f}) = {se_ind:.4f}")
print(f"Critical value z_0.05 = {z_crit:.4f}  (upper-tailed: reject if z > {z_crit:.4f})")
```

## Part 2(a): $\bar{x}_1 = 50$, $\bar{x}_2 = 60$

```python
diff_a = 50 - 60
z_a = (diff_a - D0) / se_ind
p_za = norm.sf(z_a)   # upper-tailed: P(Z > z)

print(f"x̄1 - x̄2 = {diff_a}")
print(f"z = {diff_a}/{se_ind:.4f} = {z_a:.4f}")
print(f"Upper-tailed p-value = {p_za:.4f}")
print("Reject H0" if z_a > z_crit else "Do not reject H0")
```

**Conclusion 2(a).** $z \approx -1.04 < z_{0.05} = 1.645$ and $p \approx 0.850 > 0.05$, so we **do not reject $H_0$**. As in Part 1, the sample difference ($-10$) points in the opposite direction from $H_1: \mu_1-\mu_2>0$, so these data cannot support the claim that process 1 produces more.

## Part 2(b): $\bar{x}_1 - \bar{x}_2 = 20$

```python
diff_b = 20
z_b = (diff_b - D0) / se_ind
p_zb = norm.sf(z_b)   # upper-tailed: P(Z > z)

print(f"x̄1 - x̄2 = {diff_b}")
print(f"z = {diff_b}/{se_ind:.4f} = {z_b:.4f}")
print(f"Upper-tailed p-value = {p_zb:.4f}")
print("Reject H0" if z_b > z_crit else "Do not reject H0")
```

**Conclusion 2(b).** $z \approx 2.07 > z_{0.05} = 1.645$ and $p \approx 0.019 < 0.05$, so we **reject $H_0$**. At the 5% level there is evidence that process 1 has a higher mean number of units produced per hour than process 2.

| Case | $\bar{x}_1-\bar{x}_2$ | SE | $z$ | $p$ | Decision |
|---|---|---|---|---|---|
| 2(a) | $-10$ | $9.651$ | $-1.036$ | $0.850$ | Do not reject $H_0$ |
| 2(b) | $20$ | $9.651$ | $2.072$ | $0.019$ | Reject $H_0$ |

# Part 3: Independent samples, unknown but equal variances (pooled $t$-test)

When the population variances are unknown but assumed **equal**, we pool the two sample variances:

$$
s_p^2 = \frac{(n_1 - 1)s_1^2 + (n_2 - 1)s_2^2}{n_1 + n_2 - 2}, \qquad
t = \frac{(\bar{x}_1 - \bar{x}_2) - D_0}{s_p\sqrt{\dfrac{1}{n_1} + \dfrac{1}{n_2}}}, \qquad df = n_1 + n_2 - 2
$$

For an upper-tailed test, reject $H_0$ if $t > t_{n_1+n_2-2,\alpha}$.

## Problem

You have been asked to determine if two different production processes have different mean numbers of units produced per hour. Process 1 has mean $\mu_1$ and process 2 has mean $\mu_2$. The hypotheses are:

$$
H_0: \mu_1 - \mu_2 \le 0 \qquad \text{versus} \qquad H_1: \mu_1 - \mu_2 > 0
$$

The population variances are unknown but assumed equal. A random sample of $n_1 = 25$ observations from process 1 gives $\bar{x}_1 = 56$ and $s_1 = 30$; a random sample of $n_2 = 36$ observations from process 2 gives $\bar{x}_2 = 50$ and $s_2 = 28$. Can you reject the null hypothesis using a probability of Type I error $\alpha = 0.05$?

```python
n1 = 25
n2 = 36
x_bar1 = 56
x_bar2 = 50
s1 = 30
s2 = 28
alpha = 0.05

df_pool = n1 + n2 - 2
sp2 = ((n1 - 1) * s1**2 + (n2 - 1) * s2**2) / df_pool
sp = np.sqrt(sp2)
se_pool = sp * np.sqrt(1 / n1 + 1 / n2)
t_pool = (x_bar1 - x_bar2) / se_pool
t_crit_pool = stats.t.ppf(1 - alpha, df=df_pool)
p_pool = stats.t.sf(t_pool, df=df_pool)   # upper-tailed

print(f"Pooled variance s_p² = {sp2:.4f}  ->  s_p = {sp:.4f}")
print(f"SE = s_p·√(1/{n1} + 1/{n2}) = {se_pool:.4f}")
print(f"t = ({x_bar1} - {x_bar2})/{se_pool:.4f} = {t_pool:.4f},  df = {df_pool}")
print(f"Critical value t_(0.05, {df_pool}) = {t_crit_pool:.4f}")
print(f"Upper-tailed p-value = {p_pool:.4f}")
print("Reject H0" if t_pool > t_crit_pool else "Do not reject H0")
```

**Conclusion 3.** $s_p^2 \approx 831.19$ ($s_p \approx 28.83$), SE $\approx 7.506$, $t \approx 0.80 < t_{59,0.05} \approx 1.671$, and $p \approx 0.214 > 0.05$. We **do not reject $H_0$**: a 6-unit difference in sample means is small compared with the variability, so there is not enough evidence that process 1 has a higher mean output.

# Parts 4–5: Independent samples, unknown and possibly unequal variances (Welch $t$-test)

When the population variances are unknown and **not** assumed equal, use

$$
t = \frac{(\bar{x}_1 - \bar{x}_2) - D_0}{\sqrt{\dfrac{s_1^2}{n_1} + \dfrac{s_2^2}{n_2}}}
$$

with the Welch–Satterthwaite degrees of freedom

$$
v = \frac{\left(\dfrac{s_1^2}{n_1} + \dfrac{s_2^2}{n_2}\right)^2}{\dfrac{(s_1^2/n_1)^2}{n_1 - 1} + \dfrac{(s_2^2/n_2)^2}{n_2 - 1}}
$$

For an upper-tailed test, reject $H_0$ if $t > t_{v,\alpha}$; the $p$-value is `stats.t.sf(t, df=v)`. Neither problem states a significance level, so we use $\alpha = 0.05$ and also report the $p$-value. The pooled $t$-test from Part 3 is shown as an optional check.

## Part 4: Attitudes toward minorities as managers

A screening procedure was designed to measure attitudes toward minorities as managers. High scores indicate negative attitudes and low scores indicate positive attitudes. Independent random samples were taken of $n_M = 151$ male financial analysts and $n_F = 108$ female financial analysts. For the male group the sample mean and standard deviation of scores were $\bar{x}_M = 85.8$ and $s_M = 19.13$; for the female group they were $\bar{x}_F = 71.5$ and $s_F = 12.2$.

Test the null hypothesis that the two population means are equal against the alternative that the true mean score is higher for male than for female financial analysts:

$$
H_0: \mu_M - \mu_F = 0 \qquad \text{versus} \qquad H_1: \mu_M - \mu_F > 0
$$

```python
n_M = 151
x_bar_M = 85.8
s_M = 19.13
n_F = 108
x_bar_F = 71.5
s_F = 12.2
alpha = 0.05

# Welch t-test
a_M = s_M**2 / n_M
a_F = s_F**2 / n_F
se_w = np.sqrt(a_M + a_F)
t_w = (x_bar_M - x_bar_F) / se_w
df_w = (a_M + a_F)**2 / (a_M**2 / (n_M - 1) + a_F**2 / (n_F - 1))
t_crit_w = stats.t.ppf(1 - alpha, df=df_w)
p_w = stats.t.sf(t_w, df=df_w)   # upper-tailed

print("Welch t-test")
print(f"  x̄_M - x̄_F = {x_bar_M - x_bar_F:.4f}")
print(f"  SE = √({a_M:.4f} + {a_F:.4f}) = {se_w:.4f}")
print(f"  t = {t_w:.4f},  df ≈ {df_w:.2f}")
print(f"  Critical value t_(0.05, df) = {t_crit_w:.4f}")
print(f"  Upper-tailed p-value = {p_w:.3e}")
print("  Reject H0" if t_w > t_crit_w else "  Do not reject H0")

# Optional check: pooled t-test (assumes equal variances)
df_p = n_M + n_F - 2
sp = np.sqrt(((n_M - 1) * s_M**2 + (n_F - 1) * s_F**2) / df_p)
t_p = (x_bar_M - x_bar_F) / (sp * np.sqrt(1 / n_M + 1 / n_F))
p_p = stats.t.sf(t_p, df=df_p)
print("Pooled t-test (optional)")
print(f"  s_p = {sp:.4f},  t = {t_p:.4f},  df = {df_p},  p = {p_p:.3e}")
```

**Conclusion 4.** Welch: $t \approx 7.33$ with $df \approx 254$, critical value $\approx 1.651$, $p \approx 1.5\times10^{-12}$. We **reject $H_0$** at $\alpha = 0.05$ (and at any usual level). There is very strong evidence that male financial analysts have a higher mean score, i.e. more negative attitudes toward minorities as managers, than female financial analysts. The pooled test gives the same conclusion ($t \approx 6.84$, $df = 257$, $p \approx 3\times10^{-11}$).

## Part 5: Job changes — entrepreneurs vs corporate managers

For a random sample of $n_E = 125$ British entrepreneurs, the mean number of job changes was $\bar{x}_E = 1.91$ and the sample standard deviation was $s_E = 1.32$. For an independent random sample of $n_M = 86$ British corporate managers, the mean number of job changes was $\bar{x}_M = 0.21$ and the sample standard deviation was $s_M = 0.53$.

Test the null hypothesis that the population means are equal against the alternative that the mean number of job changes is higher for British entrepreneurs than for British corporate managers:

$$
H_0: \mu_E - \mu_M = 0 \qquad \text{versus} \qquad H_1: \mu_E - \mu_M > 0
$$

```python
n_E = 125
x_bar_E = 1.91
s_E = 1.32
n_M = 86
x_bar_M = 0.21
s_M = 0.53
alpha = 0.05

# Welch t-test
a_E = s_E**2 / n_E
a_M = s_M**2 / n_M
se_w = np.sqrt(a_E + a_M)
t_w = (x_bar_E - x_bar_M) / se_w
df_w = (a_E + a_M)**2 / (a_E**2 / (n_E - 1) + a_M**2 / (n_M - 1))
t_crit_w = stats.t.ppf(1 - alpha, df=df_w)
p_w = stats.t.sf(t_w, df=df_w)   # upper-tailed

print("Welch t-test")
print(f"  x̄_E - x̄_M = {x_bar_E - x_bar_M:.4f}")
print(f"  SE = √({a_E:.4f} + {a_M:.4f}) = {se_w:.4f}")
print(f"  t = {t_w:.4f},  df ≈ {df_w:.2f}")
print(f"  Critical value t_(0.05, df) = {t_crit_w:.4f}")
print(f"  Upper-tailed p-value = {p_w:.3e}")
print("  Reject H0" if t_w > t_crit_w else "  Do not reject H0")

# Optional check: pooled t-test (assumes equal variances)
df_p = n_E + n_M - 2
sp = np.sqrt(((n_E - 1) * s_E**2 + (n_M - 1) * s_M**2) / df_p)
t_p = (x_bar_E - x_bar_M) / (sp * np.sqrt(1 / n_E + 1 / n_M))
p_p = stats.t.sf(t_p, df=df_p)
print("Pooled t-test (optional)")
print(f"  s_p = {sp:.4f},  t = {t_p:.4f},  df = {df_p},  p = {p_p:.3e}")
```

**Conclusion 5.** Welch: $t \approx 12.96$ with $df \approx 175$, critical value $\approx 1.654$, $p \approx 1.2\times10^{-27}$. We **reject $H_0$**. There is overwhelming evidence that British entrepreneurs have, on average, more job changes than British corporate managers. The sample SDs are very different (1.32 vs 0.53), which is why Welch is the better choice here; the pooled test also rejects ($t \approx 11.33$, $df = 209$).

| Part | Test | $t$ | df | $p$ | Decision ($\alpha = 0.05$) |
|---|---|---|---|---|---|
| 3 | Pooled | $0.799$ | $59$ | $0.214$ | Do not reject $H_0$ |
| 4 | Welch | $7.334$ | $\approx 254$ | $1.5\times10^{-12}$ | Reject $H_0$ |
| 5 | Welch | $12.960$ | $\approx 175$ | $1.2\times10^{-27}$ | Reject $H_0$ |

# Parts 6–7: Difference between two population proportions ($z$-test)

For independent large samples, to test $H_0: P_1 - P_2 = 0$ we use the **pooled** sample proportion (because under $H_0$ the two proportions are equal):

$$
\hat{p}_0 = \frac{n_1\hat{p}_1 + n_2\hat{p}_2}{n_1 + n_2}, \qquad
z = \frac{\hat{p}_1 - \hat{p}_2}{\sqrt{\hat{p}_0(1 - \hat{p}_0)\left(\dfrac{1}{n_1} + \dfrac{1}{n_2}\right)}}
$$

| $H_1$ | Reject $H_0$ if | $p$-value |
|---|---|---|
| $P_1 - P_2 > 0$ | $z > z_\alpha$ | `norm.sf(z)` |
| $P_1 - P_2 \neq 0$ | $|z| > z_{\alpha/2}$ | `2 * norm.sf(abs(z))` |

## Part 6: Economic optimism — United States vs Great Britain

Random samples of $n_{US} = 900$ people in the United States and $n_{GB} = 900$ people in Great Britain indicated that $\hat{p}_{US} = 0.60$ of the people in the United States were positive about the future economy, whereas $\hat{p}_{GB} = 0.66$ of the people in Great Britain were positive about the future economy. Does this provide strong evidence that the people in Great Britain are more optimistic about the economy?

$$
H_0: P_{GB} - P_{US} = 0 \qquad \text{versus} \qquad H_1: P_{GB} - P_{US} > 0
$$

Use $\alpha = 0.05$ and report the $p$-value.

```python
n_GB = 900
p_GB = 0.66
n_US = 900
p_US = 0.60
alpha = 0.05

p_pool6 = (n_GB * p_GB + n_US * p_US) / (n_GB + n_US)
se6 = np.sqrt(p_pool6 * (1 - p_pool6) * (1 / n_GB + 1 / n_US))
z6 = (p_GB - p_US) / se6
z_crit6 = norm.ppf(1 - alpha)
p_value6 = norm.sf(z6)   # upper-tailed

print(f"Pooled proportion p̂0 = {p_pool6:.4f}")
print(f"SE = {se6:.5f}")
print(f"z = ({p_GB} - {p_US})/{se6:.5f} = {z6:.4f}")
print(f"Critical value z_0.05 = {z_crit6:.4f}")
print(f"Upper-tailed p-value = {p_value6:.4f}")
print("Reject H0" if z6 > z_crit6 else "Do not reject H0")
```

**Conclusion 6.** $\hat{p}_0 = 0.63$, SE $\approx 0.0228$, $z \approx 2.64 > 1.645$, and $p \approx 0.0042$. We **reject $H_0$** at the 5% level; in fact we would also reject at the 1% level. Yes, this is strong evidence that people in Great Britain are more optimistic about the economy than people in the United States.

## Part 7 (10.16): Attitudes toward world trade — country A vs country B

A random sample of $n_A = 1556$ people in country A were asked to respond to this statement: *Increased world trade can increase our per capita prosperity.* Of these sample members, $\hat{p}_A = 0.384$ (38.4%) agreed with the statement. When the same statement was presented to a random sample of $n_B = 1108$ people in country B, $\hat{p}_B = 0.520$ (52.0%) agreed. Test the null hypothesis that the population proportions agreeing with this statement were the same in the two countries against the two-sided alternative:

$$
H_0: P_A - P_B = 0 \qquad \text{versus} \qquad H_1: P_A - P_B \neq 0
$$

Use $\alpha = 0.05$ and report the $p$-value.

```python
n_A = 1556
p_A = 0.384
n_B = 1108
p_B = 0.520
alpha = 0.05

p_pool7 = (n_A * p_A + n_B * p_B) / (n_A + n_B)
se7 = np.sqrt(p_pool7 * (1 - p_pool7) * (1 / n_A + 1 / n_B))
z7 = (p_A - p_B) / se7
z_crit7 = norm.ppf(1 - alpha / 2)
p_value7 = 2 * norm.sf(abs(z7))   # two-sided

print(f"Pooled proportion p̂0 = {p_pool7:.4f}")
print(f"SE = {se7:.5f}")
print(f"z = ({p_A} - {p_B})/{se7:.5f} = {z7:.4f}")
print(f"Critical values = ±{z_crit7:.4f}")
print(f"Two-sided p-value = {p_value7:.3e}")
print("Reject H0" if abs(z7) > z_crit7 else "Do not reject H0")
```

**Conclusion 7.** $\hat{p}_0 \approx 0.4406$, SE $\approx 0.0195$, $z \approx -6.97$, so $|z| > 1.96$, and the two-sided $p \approx 3.2\times10^{-12}$. We **reject $H_0$**: the proportion agreeing with the statement differs between the two countries, and it is clearly lower in country A.

| Part | $\hat{p}_1 - \hat{p}_2$ | $\hat{p}_0$ | SE | $z$ | $p$ | Decision ($\alpha=0.05$) |
|---|---|---|---|---|---|---|
| 6 (GB − US, upper-tailed) | $0.060$ | $0.630$ | $0.0228$ | $2.636$ | $0.0042$ | Reject $H_0$ |
| 7 (A − B, two-sided) | $-0.136$ | $0.4406$ | $0.0195$ | $-6.969$ | $3.2\times10^{-12}$ | Reject $H_0$ |

# Parts 8–9: Equality of two population variances ($F$-test)

For independent random samples from two **normal** populations, to test $H_0: \sigma_x^2 = \sigma_y^2$ against $H_1: \sigma_x^2 > \sigma_y^2$ we use

$$
F = \frac{s_x^2}{s_y^2}, \qquad \text{with } (n_x - 1,\; n_y - 1) \text{ degrees of freedom}
$$

where the variance under the alternative's "larger" side goes in the **numerator**. Reject $H_0$ if $F > F_{n_x-1,\,n_y-1,\,\alpha}$; the $p$-value is $P(F_{n_x-1,n_y-1} > F)$, computed with `stats.f.sf(F, dfn, dfd)`.

## Part 8: Variability of tax-filing judgments by expertise

It is hypothesized that the more expert a group of people examining personal income tax filings, the more variable the judgments will be about the accuracy. Independent random samples, each of 30 individuals, were chosen from groups with different levels of expertise. The low-expertise group consisted of people who had just completed their first intermediate accounting course. Members of the high-expertise group had completed undergraduate studies and were employed by reputable CPA firms. The sample members were asked to judge the accuracy of personal income tax filings. For the low-expertise group the sample variance was $s_L^2 = 451.770$, whereas for the high-expertise group it was $s_H^2 = 1614.208$. Test the null hypothesis that the two population variances are equal against the alternative that the true variance is higher for the high-expertise group:

$$
H_0: \sigma_H^2 = \sigma_L^2 \qquad \text{versus} \qquad H_1: \sigma_H^2 > \sigma_L^2
$$

No significance level is given, so we use $\alpha = 0.05$ (and also check $\alpha = 0.01$) and report the $p$-value. We assume both populations are normal.

```python
n_H = 30
var_H = 1614.208
n_L = 30
var_L = 451.770
alpha = 0.05

F8 = var_H / var_L
dfn8 = n_H - 1
dfd8 = n_L - 1
F_crit8 = stats.f.ppf(1 - alpha, dfn8, dfd8)
F_crit8_01 = stats.f.ppf(1 - 0.01, dfn8, dfd8)
p8 = stats.f.sf(F8, dfn8, dfd8)   # upper-tailed

print(f"F = {var_H}/{var_L} = {F8:.4f},  df = ({dfn8}, {dfd8})")
print(f"Critical value F_(29, 29, 0.05) = {F_crit8:.4f}")
print(f"Critical value F_(29, 29, 0.01) = {F_crit8_01:.4f}")
print(f"Upper-tailed p-value = {p8:.5f}")
print("Reject H0 at 5%" if F8 > F_crit8 else "Do not reject H0 at 5%")
```

**Conclusion 8.** $F \approx 3.573$ with $(29, 29)$ df, compared with $F_{29,29,0.05} \approx 1.861$ (and $F_{29,29,0.01} \approx 2.423$), and $p \approx 0.0005$. We **reject $H_0$** at the 5% level and also at 1%. There is strong evidence that judgments of the high-expertise group are more variable than those of the low-expertise group, which supports the hypothesis.

## Part 9 (10.24): Variability of ship sales — price competition vs duopoly

It is hypothesized that the total sales of a corporation should vary more in an industry with active price competition than in one with duopoly and tacit collusion. In a study of the merchant ship production industry it was found that in $n_C = 4$ years of active price competition, the variance of company A's total sales was $s_C^2 = 114.09$. In the following $n_D = 7$ years, during which there was duopoly and tacit collusion, this variance was $s_D^2 = 16.08$. Assume that the data can be regarded as an independent random sample from two normal distributions. Test, at the 5% level, the null hypothesis that the two population variances are equal against the alternative that the variance of total sales is higher in years of active price competition:

$$
H_0: \sigma_C^2 = \sigma_D^2 \qquad \text{versus} \qquad H_1: \sigma_C^2 > \sigma_D^2
$$

```python
n_C = 4
var_C = 114.09
n_D = 7
var_D = 16.08
alpha = 0.05

F9 = var_C / var_D
dfn9 = n_C - 1
dfd9 = n_D - 1
F_crit9 = stats.f.ppf(1 - alpha, dfn9, dfd9)
p9 = stats.f.sf(F9, dfn9, dfd9)   # upper-tailed

print(f"F = {var_C}/{var_D} = {F9:.4f},  df = ({dfn9}, {dfd9})")
print(f"Critical value F_(3, 6, 0.05) = {F_crit9:.4f}")
print(f"Upper-tailed p-value = {p9:.4f}")
print("Reject H0" if F9 > F_crit9 else "Do not reject H0")
```

**Conclusion 9.** $F \approx 7.095$ with $(3, 6)$ df, which exceeds $F_{3,6,0.05} \approx 4.757$, and $p \approx 0.021 < 0.05$. We **reject $H_0$** at the 5% level: the variance of total sales is higher in years of active price competition. Note that the samples are very small (4 and 7 years), and the $F$-test depends heavily on the normality assumption.

|---|---|---|---|---|---|
| 8 (high vs low expertise) | $3.573$ | $(29, 29)$ | $1.861$ | $0.0005$ | Reject $H_0$ |
| 9 (competition vs duopoly) | $7.095$ | $(3, 6)$ | $4.757$ | $0.021$ | Reject $H_0$ |

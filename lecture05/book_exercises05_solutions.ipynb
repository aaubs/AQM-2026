### Formulas from sections 1.2, 2.1 and 2.2

- Quartiles (Newbold): $Q_1$, $Q_2$, $Q_3$ are the values in position $0.25(n+1)$, $0.50(n+1)$ and $0.75(n+1)$ of the ordered data. If the position is not a whole number, interpolate between the two neighbours. $\text{IQR} = Q_3 - Q_1$.
- Coefficient of variation: $CV = s/\bar{x}$. $z$-score: $z = (x_i - \bar{x})/s$.
- Chebyshev (any distribution), $k > 1$: at least $1 - 1/k^2$ of the observations lie within $k$ standard deviations of the mean.
- Empirical rule (mound-shaped distribution): about 68%, 95% and almost all observations lie within 1, 2 and 3 standard deviations.
- Three kinds of missing data: completely at random, at random (depends on something we observe), not at random (depends on the missing value itself).

### Test statistics

| Test | Statistic | Distribution under $H_0$ |
|---|---|---|
| Paired $t$ | $t = \bar d / (s_d/\sqrt{n})$ | $t$ with $n-1$ df |
| Two means, variances known | $z = (\bar x_1 - \bar x_2)/\sqrt{\sigma_1^2/n_1 + \sigma_2^2/n_2}$ | standard normal |
| Two means, pooled | $t = (\bar x_1 - \bar x_2)/\big(s_p\sqrt{1/n_1 + 1/n_2}\big)$ | $t$ with $n_1+n_2-2$ df |
| Two means, Welch | $t = (\bar x_1 - \bar x_2)/\sqrt{s_1^2/n_1 + s_2^2/n_2}$ | $t$ with Welch $v$ df |
| Two proportions | $z = (\hat p_1 - \hat p_2)/\sqrt{\hat p_0(1-\hat p_0)(1/n_1+1/n_2)}$ | approx. standard normal |
| Two variances | $F = s_1^2/s_2^2$ (larger variance on top for one-sided) | $F$ with $n_1-1$, $n_2-1$ df |

### Task 2. A survey that few late customers answer

RetailCorp invites all 1,500 customers who ordered in June to rate their delivery from 1 to 5. Of these, 300 had a late delivery and 1,200 did not. If everyone answered, the late customers would give an average rating of 2.5 and the others 4.4. In reality 25% of the late customers answer, against 50% of the others. Assume that the customers who answer in each group are typical of their group.

a. What is the true mean rating among all 1,500 customers?  
b. How many customers answer in each group, and what is the mean rating among everyone who answers?  
c. By how much is the survey mean off? Is this a sampling error or a nonsampling error, and would inviting twice as many customers (with the same response rates) remove it?  
d. Suppose you know that 20% of all customers had a late delivery. Weight the two group means by 20% and 80%. What do you get, and what must be true for this to work?

<details>
<summary>Hint 1</summary>

The mean of everyone is the mean of the group means, weighted by group size. The mean of the responders weights by the number of responders in each group. Sampling error shrinks when $n$ grows; a bias from who answers does not.
</details>

<details>
<summary>Hint 2</summary>

Responders in a group $=$ invited $\times$ response rate. Weighted mean $=(n_1 m_1 + n_2 m_2)/(n_1 + n_2)$. Plain arithmetic in Python is enough.
</details>

```python
n_late, n_ok = 300, 1200
mean_late, mean_ok = 2.5, 4.4
rate_late, rate_ok = 0.25, 0.50

true_mean = (n_late * mean_late + n_ok * mean_ok) / (n_late + n_ok)

resp_late, resp_ok = n_late * rate_late, n_ok * rate_ok
survey_mean = (resp_late * mean_late + resp_ok * mean_ok) / (resp_late + resp_ok)

weighted_mean = 0.20 * mean_late + 0.80 * mean_ok

print(f"True mean rating of all customers: {true_mean:.3f}")
print(f"Responders: {resp_late:.0f} late, {resp_ok:.0f} not late")
print(f"Mean among responders:             {survey_mean:.3f}")
print(f"Survey minus true mean:            {survey_mean - true_mean:+.3f}")
print(f"Mean weighted by 20% / 80%:        {weighted_mean:.3f}")
```

**Answer.** The true mean is 4.02. 75 late and 600 other customers answer, and their mean is 4.19, which is 0.17 too high because late (unhappy) customers answer less often. This is a nonsampling error (section 1.1: the group that answers is not the population of interest). Inviting twice as many customers gives the same bias with less noise around it. Weighting the group means by the true shares gives back 4.02, but only because the customers who answer in each group are typical of their group. If, for example, only the angriest late customers answered, no weighting on data we can see would correct it.

### Task 3. Weekly complaints and a code for "not recorded"

The support desk counted complaints about late deliveries in nine randomly chosen weeks: 14, 17, 15, 16, 14, 18, 15, 99, 14. The data description says that 99 is the code for a week that was not recorded.

a. Compute the mean, median and mode of the nine values as they stand. Which measure describes a typical week best, and why?  
b. Replace the code by a missing value and compute the three measures again.  
c. Which measure changed most? Would Python have warned you about the code if you had not read the data description?

<details>
<summary>Hint 1</summary>

The mean uses every value, so one extreme value moves it. The median is the middle value of the ordered data and the mode is the most frequent value. To Python, 99 is just a number until you replace it.
</details>

<details>
<summary>Hint 2</summary>

`pd.Series([...])`, `.mean()`, `.median()`, `.mode()` (a Series), and `x.replace(99, np.nan)`. pandas skips `NaN` in all three. Use `.count()` to see how many values are left.
</details>

```python
weeks = pd.Series([14, 17, 15, 16, 14, 18, 15, 99, 14])
cleaned = weeks.replace(99, np.nan)

for name, x in (("as they stand", weeks), ("99 set to missing", cleaned)):
    print(f"{name:18s} n = {x.count()}, mean = {x.mean():.2f}, "
          f"median = {x.median():.1f}, mode = {x.mode().tolist()}")
```

**Answer.** As they stand: mean 24.67, median 15, mode 14. With 99 set to missing (n = 8): mean 15.38, median 15, mode 14. Only the mean changes (by more than 9 complaints), because one extreme value pulls it; the median and the mode do not move. For the raw data the median describes a typical week best. Python would not have warned you: `mean()` happily includes 99 and gives a wrong number without any error.

### Task 4. Quartiles by Newbold's rule, and by pandas

The returns desk counted the parcels returned on 12 randomly chosen days: 23, 31, 27, 35, 29, 26, 41, 24, 30, 28, 33, 25.

a. Find the five-number summary with Newbold's rule: sort the data and take the values in position $0.25(n+1)$, $0.50(n+1)$ and $0.75(n+1)$, interpolating when the position is not a whole number. Then check your numbers with `np.percentile(..., method="weibull")`.  
b. Compute the interquartile range.  
c. Compute the quartiles with the default rule of pandas and NumPy (for example with `describe()`). Which numbers differ from (a)?  
d. A common rule flags a value more than $1.5 \times \text{IQR}$ above $Q_3$. Is 41 flagged with the quartiles from (a)? With those from (c)? What does that tell you about a flag?

<details>
<summary>Hint 1</summary>

Sort first. For $n = 12$ the positions are 3.25, 6.5 and 9.75. For a position like 3.25, take the 3rd ordered value plus 0.25 of the gap to the 4th. The IQR is $Q_3 - Q_1$.
</details>

<details>
<summary>Hint 2</summary>

`np.sort`, indexing (position 3 is index 2), `np.percentile(x, [25, 50, 75], method=\"weibull\")` for Newbold's rule, and `pd.Series(x).describe()` or `np.percentile(x, [25, 50, 75])` for the default.
</details>

```python
returns = np.array([23, 31, 27, 35, 29, 26, 41, 24, 30, 28, 33, 25])
ordered = np.sort(returns)

def newbold_quantile(ordered, p):
    pos = p * (len(ordered) + 1)              # 1-based position
    lower = int(np.floor(pos))
    return ordered[lower - 1] + (pos - lower) * (ordered[lower] - ordered[lower - 1])

q1, q2, q3 = (newbold_quantile(ordered, p) for p in (0.25, 0.50, 0.75))
print("Five-number summary (Newbold):", ordered.min(), q1, q2, q3, ordered.max())
print("Check with np.percentile:     ", np.percentile(returns, [25, 50, 75], method="weibull"))
print(f"IQR = {q3 - q1}")

pq1, pq2, pq3 = np.percentile(returns, [25, 50, 75])      # the default rule
print("Default rule:", pq1, pq2, pq3, f" IQR = {pq3 - pq1}")
print(pd.Series(returns).describe())

for name, a, b in (("Newbold", q1, q3), ("default", pq1, pq3)):
    fence = b + 1.5 * (b - a)
    print(f"{name}: fence = {fence:.3f}, 41 flagged: {41 > fence}")
```

**Answer.** Newbold's rule gives 23, 25.25, 28.5, 32.5, 41, so IQR = 7.25. The default rule gives $Q_1 = 25.75$ and $Q_3 = 31.5$ (IQR = 5.75); the median is the same. The default places quartiles at position $p(n-1)+1$ instead of $p(n+1)$, so with 12 values they sit slightly closer to the middle. The upper fence is 43.4 with Newbold's quartiles, so 41 is not flagged, but 40.1 with the default quartiles, so 41 is flagged. A flag is therefore a reason to look at the value (a day with 41 returns can be real), not a verdict, and it can depend on how the quartiles are defined.

### Task 5. Two routes with the same mean

A carrier compares the delivery times (in hours) on two routes, with 10 randomly chosen deliveries on each.

Route A: 34, 36, 35, 38, 33, 37, 35, 36, 34, 37  
Route B: 28, 42, 31, 39, 26, 44, 35, 33, 40, 37

a. Compute the mean and the median of each route.  
b. Compute the range, the IQR (Newbold's rule), the sample standard deviation and the coefficient of variation of each route.  
c. Draw a box plot of the two routes side by side.  
d. The carrier says "both routes take about 35.5 hours, so they are equally good". Do you agree? Which measure of variability would you report, and why is the mean alone not enough?

<details>
<summary>Hint 1</summary>

The same mean does not mean the same spread. The range uses only the two extreme values, the IQR the middle 50% of the data, and $s$ and $CV = s/\bar{x}$ use every value. Use $n - 1$ in $s$.
</details>

<details>
<summary>Hint 2</summary>

`x.mean()`, `np.median(x)`, `x.max() - x.min()`, `np.percentile(x, [25, 75], method=\"weibull\")`, `x.std(ddof=1)`; `plt.boxplot([a, b])` and `plt.xticks([1, 2], [\"A\", \"B\"])`.
</details>

```python
route_a = np.array([34, 36, 35, 38, 33, 37, 35, 36, 34, 37])
route_b = np.array([28, 42, 31, 39, 26, 44, 35, 33, 40, 37])

rows = []
for name, x in (("A", route_a), ("B", route_b)):
    q1, q3 = np.percentile(x, [25, 75], method="weibull")
    rows.append({"route": name, "mean": x.mean(), "median": np.median(x),
                 "range": x.max() - x.min(), "IQR": q3 - q1,
                 "s": x.std(ddof=1), "CV": x.std(ddof=1) / x.mean()})
print(pd.DataFrame(rows).round(3).to_string(index=False))

plt.boxplot([route_a, route_b])
plt.xticks([1, 2], ["Route A", "Route B"])
plt.ylabel("Delivery time (hours)")
plt.show()
```

**Answer.** Both routes have mean 35.5 (medians 35.5 and 36). Route A has range 5, IQR 3.0, $s = 1.58$ and $CV = 0.044$; route B has range 18, IQR 10.25, $s = 5.99$ and $CV = 0.169$. Route B is about four times as variable, so the routes are not equally good: on route B a customer can wait anywhere from about 26 to 44 hours. Report the mean (or median) together with $s$ or the IQR; the IQR is the safer choice if there may be extreme values, because it ignores the tails.

### Task 6. "Don't know": drop or fill?

A short delivery survey asked 14 randomly chosen customers to rate their delivery from 1 (very poor) to 5 (very good). The answers were: 5, 4, −1, 4, 5, 3, 4, 5, −1, 2, 4, 5, 4, −1. The code −1 means "don't know".

a. Compute the minimum, mean and standard deviation of the 14 values as they stand. Does the output show that something is wrong?  
b. Replace the code by a missing value. How many ratings are missing, and what are the mean, median and standard deviation of the rest?  
c. Fill the missing values with the mean of the rest. What are the new mean and standard deviation? Compare with (b).  
d. Compute a 95% confidence interval for the mean rating (as in Lecture 2) from the data with the missing values dropped and from the filled data. Which interval is wider, and which is honest?

<details>
<summary>Hint 1</summary>

−1 is outside the 1 to 5 scale, so it is a code and not an answer. Dropping the rows keeps $n$ honest; filling with the mean keeps $n = 14$, but every filled value sits exactly at the mean, which pulls $s$ down.
</details>

<details>
<summary>Hint 2</summary>

`x.replace(-1, np.nan)`, `x.isna().sum()`, `x.fillna(x.mean())`, and `stats.t.interval(0.95, n - 1, loc=x.mean(), scale=x.std(ddof=1) / np.sqrt(n))`. For the dropped data use $n$ = the number of ratings left.
</details>

```python
ratings = pd.Series([5, 4, -1, 4, 5, 3, 4, 5, -1, 2, 4, 5, 4, -1])
print(f"As they stand: min {ratings.min()}, mean {ratings.mean():.3f}, s {ratings.std():.3f}")

clean = ratings.replace(-1, np.nan)
print(f"Missing: {clean.isna().sum()}; mean {clean.mean():.3f}, "
      f"median {clean.median():.1f}, s {clean.std():.3f}")

filled = clean.fillna(clean.mean())
print(f"Filled:  mean {filled.mean():.3f}, s {filled.std():.3f}")

for name, x in (("dropped", clean.dropna()), ("filled", filled)):
    n = len(x)
    lo, hi = stats.t.interval(0.95, n - 1, loc=x.mean(), scale=x.std(ddof=1) / np.sqrt(n))
    print(f"{name:8s} n = {n}, 95% interval [{lo:.3f}, {hi:.3f}], width {hi - lo:.3f}")
```

**Answer.** As they stand, the mean is 3.00 and $s = 2.32$; only the minimum of −1 shows that something is wrong, and the mean looks perfectly plausible. After replacing the code, 3 ratings are missing and the other 11 have mean 4.09, median 4 and $s = 0.94$. Filling with the mean keeps the mean at 4.09 but lowers $s$ to 0.83. The interval from the 11 real ratings is [3.46, 4.73] (width 1.27); the interval from the filled data is [3.61, 4.57] (width 0.96). The narrower one is not more precise: it counts three invented ratings as if they were information. The dropped-data interval is the honest one, and it describes the customers who gave a rating.

### Problem 2. Chebyshev, the empirical rule and z-scores

The delivery times of a carrier have a mean of 36 hours and a standard deviation of 8 hours.

1. If you know nothing about the shape of the distribution, what share of the deliveries is guaranteed to take between 20 and 52 hours?
2. If the distribution is mound-shaped, about what share takes between 28 and 44 hours, and about what share between 20 and 52 hours?
3. Compute the $z$-scores of deliveries that took 62, 14 and 40 hours. Which of them would the 3 $s$ rule from the lecture flag as an outlier?

<details>
<summary>Hint 1</summary>

Find how many standard deviations the end points are from the mean: $k = (x - \bar{x})/s$. Chebyshev: at least $1 - 1/k^2$ for any distribution. Empirical rule: about 68%, 95% and 99.7% for $k = 1, 2, 3$.
</details>

<details>
<summary>Hint 2</summary>

Plain arithmetic, with the times in a NumPy array: `z = (times - mean) / sd`, and `np.abs(z) > 3` picks the flagged ones.
</details>

```python
mean, sd = 36, 8
k = (52 - mean) / sd
print(f"k = {k}: Chebyshev guarantees at least {1 - 1 / k**2:.0%} between {mean - k*sd} and {mean + k*sd} hours")
print(f"Empirical rule: about 68% between {mean - sd} and {mean + sd}, about 95% between {mean - 2*sd} and {mean + 2*sd}")

times = np.array([62, 14, 40])
z = (times - mean) / sd
print(dict(zip(times.tolist(), z.round(2).tolist())))
print("Flagged by the 3 s rule:", times[np.abs(z) > 3].tolist())
```

**Answer.** 20 and 52 are $k = 2$ standard deviations from the mean, so Chebyshev guarantees at least $1 - 1/4 = 75\%$. For a mound-shaped distribution about 68% take between 28 and 44 hours and about 95% between 20 and 52 hours. The $z$-scores are 3.25 (62 hours), −2.75 (14 hours) and 0.5 (40 hours), so only the 62-hour delivery is flagged. Note that the same interval, 20 to 52 hours, holds at least 75% for any distribution but about 95% if the distribution is mound-shaped.

### Problem 3. How wide an interval is guaranteed?

Invoice amounts have a mean of 1,240 DKK and a standard deviation of 310 DKK. We do not know that the distribution is bell-shaped.

1. Between which two amounts are at least 75% of the invoices guaranteed to lie?
2. Between which two amounts are at least 90% guaranteed to lie? (Solve $1 - 1/k^2 = 0.90$ for $k$.)
3. If the distribution were bell-shaped, which interval would hold about 95% of the invoices? Why is the guaranteed interval for 90% so much wider than the 95% interval from the empirical rule?

<details>
<summary>Hint 1</summary>

Chebyshev: at least $1 - 1/k^2$ of the values lie within $k$ standard deviations of the mean. Solve for $k = 1/\sqrt{1 - p}$ and use $\bar{x} \pm k s$.
</details>

<details>
<summary>Hint 2</summary>

`np.sqrt` and plain arithmetic; a loop over the two shares keeps the code short.
</details>

```python
mean, sd = 1240, 310
for p in (0.75, 0.90):
    k = 1 / np.sqrt(1 - p)
    print(f"At least {p:.0%}: k = {k:.3f}, from {mean - k*sd:.0f} to {mean + k*sd:.0f} DKK")
print(f"Empirical rule, about 95%: {mean - 2*sd} to {mean + 2*sd} DKK")
```

**Answer.** At least 75% lie between 620 and 1,860 DKK ($k = 2$). For 90%, $k = 1/\sqrt{0.10} = 3.16$, so the interval is about 260 to 2,220 DKK. If the distribution were bell-shaped, about 95% would lie in the interval 620 to 1,860 DKK, the same interval for which Chebyshev only guarantees 75%. Chebyshev has to hold for every distribution, including very unfavourable ones, so its guarantees are cautious and its intervals wide.

### Extension 1. The empirical rule on skewed data

Order values are right-skewed. Simulate 10,000 order values with `rng = np.random.default_rng(5)` and `orders = rng.lognormal(mean=6, sigma=0.5, size=10_000)` (DKK).

1. Compute the mean, median, standard deviation and skewness. What do the gap between mean and median and the skewness tell you?
2. What share of the values lies within 1, 2 and 3 standard deviations of the mean? Compare with the empirical rule (68%, 95%, 99.7%) and with Chebyshev's guarantees.
3. How many values lie more than $1.5 \times \text{IQR}$ above $Q_3$ (use Newbold's quartiles)? Are these errors?

<details>
<summary>Hint 1</summary>

A mean above the median and a positive skewness signal a long right tail. Count the share inside $\bar{x} \pm k s$ with a condition on both sides and `.mean()`.
</details>

<details>
<summary>Hint 2</summary>

`rng.lognormal(mean=6, sigma=0.5, size=10_000)`, `.mean()`, `np.median`, `.std(ddof=1)`, `stats.skew`, `((x > lo) & (x < hi)).mean()` and `np.percentile(x, [25, 75], method=\"weibull\")`.
</details>

```python
rng = np.random.default_rng(5)
orders = rng.lognormal(mean=6, sigma=0.5, size=10_000)
m, s = orders.mean(), orders.std(ddof=1)
print(f"mean {m:.1f}, median {np.median(orders):.1f}, s {s:.1f}, skewness {stats.skew(orders):.2f}")

for k, rule in zip((1, 2, 3), (68, 95, 99.7)):
    share = ((orders > m - k * s) & (orders < m + k * s)).mean()
    print(f"Within {k} s: {share:.1%}  (empirical rule {rule}%, Chebyshev at least {max(0, 1 - 1 / k**2):.1%})")

q1, q3 = np.percentile(orders, [25, 75], method="weibull")
fence = q3 + 1.5 * (q3 - q1)
print(f"Values above {fence:.0f}: {(orders > fence).sum()} of {len(orders)}")
```

**Answer.** The mean (about 463) is well above the median (about 408) and the skewness is 1.68: a long right tail. The shares within 1, 2 and 3 standard deviations are 76.0%, 95.4% and 98.7%. The empirical rule is wrong at both ends: too few values (68%) are expected inside 1 $s$, because many values bunch up below the mean, and too many (99.7%) inside 3 $s$, because the tail is heavy. Chebyshev's guarantees (none for $k = 1$, 75%, 88.9%) hold, as they must. 363 values (3.6%) lie above the fence at about 1,003 DKK. They are not errors: the data are simulated from a skewed distribution, so a skewed file will always have many large but genuine values. A flag is a reason to look at the order, not to delete it.

### Extension 2. How much does one wrong value move each measure?

Ten delivery times in hours: 33, 34, 34, 35, 35, 36, 36, 37, 37, 38. Suppose the last value, 38, is entered wrongly in one of four ways: 38 (correct), 80, 380 (the decimal point was lost) or 3,800.

1. For each version compute the mean, median, sample standard deviation, range and IQR (Newbold's rule).
2. Which measures barely move, and which explode?
3. What does this tell you about which measures to trust before the data are cleaned, and what must you still do?

<details>
<summary>Hint 1</summary>

The mean, the standard deviation and the range use the extreme value directly. The median and the IQR depend only on the order of the values, so one extreme value moves them little or not at all.
</details>

<details>
<summary>Hint 2</summary>

A `for` loop over the four values: copy the array (`x = base.copy()`), replace the last element (`x[-1] = last`), compute the measures, append a dictionary to a list and build a `pd.DataFrame` from the list.
</details>

```python
base = np.array([33, 34, 34, 35, 35, 36, 36, 37, 37, 38])
rows = []
for last in (38, 80, 380, 3800):
    x = base.copy()
    x[-1] = last
    q1, q3 = np.percentile(x, [25, 75], method="weibull")
    rows.append({"last value": last, "mean": x.mean(), "median": np.median(x),
                 "s": x.std(ddof=1), "range": x.max() - x.min(), "IQR": q3 - q1})
print(pd.DataFrame(rows).round(1).to_string(index=False))
```

**Answer.** The median stays at 35.5 and the IQR at 3.0 in all four versions. The mean goes from 35.5 to 39.7, 69.7 and 411.7; the standard deviation from 1.6 to 14.2, 109.0 and 1,190.5; the range from 5 to 47, 347 and 3,767. Before the data are cleaned, the median and the IQR are the safer summaries, because one wrong value hardly moves them. But a robust measure only hides the problem: you still have to look for the wrong value (a minimum and maximum, a histogram) and decide what it should be, since a lost decimal point also corrupts every analysis that uses the mean.

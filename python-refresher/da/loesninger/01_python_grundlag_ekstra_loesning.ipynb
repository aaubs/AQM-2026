## Blok 3: Break-even, vækst og sympy (til øvelse 3)

```python
# Opsætning til blok 3: kør denne celle først
import numpy as np
import matplotlib.pyplot as plt

# Kioskens tal pr. dag
faste = 400       # faste omkostninger i kr.
variabel = 12     # omkostning pr. solgt is i kr.
pris = 45         # salgspris pr. is i kr.
```

### Opgave 3.1: Break-even beregnet med kode

I notebook 1 fik du break-even givet som 400/33. Her skal du beregne det ud fra variablerne, så resultatet følger med, hvis tallene ændres.

Ved break-even er omsætningen lig med omkostningerne: $\text{pris} \cdot x = \text{faste} + \text{variabel} \cdot x$. Løs ligningen for $x$ på papir, og skriv formlen i koden med variablernes navne.

```python
x_be = faste / (pris - variabel)
print(f"Break-even ved {x_be:.2f} solgte is")
```

```python
#@title Tjek dit svar (kør cellen)
print("Rigtigt" if np.isclose(pris * x_be, faste + variabel * x_be) else "Ikke endnu: ved break-even skal omsætning og omkostning være lige store")
```

### Opgave 3.2: Figur med hjælpelinjer

Udfyld hullerne, så figuren viser begge kurver, markerer break-even med en prik og tegner stiplede hjælpelinjer gennem punktet. `plt.axvline(a)` tegner en lodret linje ved $x = a$, og `plt.axhline(b)` tegner en vandret linje ved $y = b$.

```python
x = np.linspace(0, 40, 100)
omkostning = faste + variabel * x
omsaetning = pris * x
y_be = pris * x_be    # kronebeløbet ved break-even

plt.plot(x, omkostning, label="Omkostning")
plt.plot(x, omsaetning, label="Omsætning")
plt.plot(x_be, y_be, "o", label="Break-even")
plt.axvline(x_be, linestyle="--", color="grey")
plt.axhline(y_be, linestyle="--", color="grey")
plt.xlabel("Antal solgte is")
plt.ylabel("Kroner")
plt.title("Ishuset: break-even for en kiosk")
plt.legend()
plt.show()
```

### Opgave 3.4: Vækst over tid

### Opgave 3.5: Break-even med sympy

Python kan også regne med symboler, som man gør i matematik. Det gør pakken sympy, som vi importerer med aliasset `sp`.

Du vil måske også se `from sympy import symbols, solve`. Så skriver man `symbols(...)` i stedet for `sp.symbols(...)`. Det er de samme funktioner.

I cellen herunder er $x$ defineret som et symbol, og omkostning og omsætning er skrevet som udtryk i $x$. Bemærk, at `x` var et numpy-array i opgave 3.2. Nu er `x` et symbol, fordi vi har givet navnet en ny værdi.

```python
import sympy as sp

x = sp.symbols("x")
C = 400 + 12 * x     # omkostning
R = 45 * x           # omsætning
print(C)
print(R)
```

`sp.Eq(C, R)` er ligningen $C = R$, og `sp.solve(ligning, x)` løser den for $x$. Udfyld hullet.

```python
loesning = sp.solve(sp.Eq(C, R), x)
print(loesning)
print(type(loesning))
```

Resultatet er en liste med én løsning, og sympy giver den som en eksakt brøk. Tag løsningen ud af listen med et indeks, og lav den om til et decimaltal med `float()`.

```python
x_be_sympy = float(loesning[0])
print(f"Break-even ved {x_be_sympy:.2f} solgte is")
```

```python
#@title Tjek dit svar (kør cellen)
print("Rigtigt" if np.isclose(x_be_sympy, 400 / 33) else "Ikke endnu: husk både indekset [0] og float()")
```

### Opgave 3.6: Optimering med sympy

Ishuset har modelleret kioskens daglige profit $P(x)$ i kr. som funktion af antal solgte is $x$, målt i tiere ($x = 30$ betyder 300 is):

$$P(x) = -0{,}01x^3 + 0{,}75x^2 + 18x - 500$$

Ved mange solgte is stiger udgifterne til ekstra personale og køleplads, så profitten falder igen. Find det antal is, der giver den største profit.

Først de afledte funktioner. `sp.diff(udtryk, x)` differentierer udtrykket med hensyn til $x$.

```python
x = sp.symbols("x")
P = -0.01 * x**3 + 0.75 * x**2 + 18 * x - 500

# Første og anden afledte
P1 = sp.diff(P, x)
P2 = sp.diff(P1, x)
print(P1)
print(P2)
```

De kritiske punkter er der, hvor $P'(x) = 0$. Giver man `sp.solve` et udtryk uden lighedstegn, løser den ligningen *udtryk = 0*.

```python
kritiske = sp.solve(P1, x)
print(kritiske)
```

Der er to kritiske punkter, ét negativt og ét positivt. Et negativt antal is giver ikke mening. Vælg det positive med et indeks (se på outputtet ovenfor for at se, om det står først eller sidst), og lav andenafledte-testen.

```python
x_opt = kritiske[1]
test = P2.subs(x, x_opt)
print(f"P''({x_opt:.0f}) = {test:.2f}")

if test < 0:
    print("Negativ: lokalt maksimum")
else:
    print("Positiv: lokalt minimum")
```

Beregn den største profit ved at indsætte `x_opt` i `P` med `.subs()`, og lav resultatet om til et decimaltal med `float()`.

```python
P_max = float(P.subs(x, x_opt))
print(f"Største profit: {P_max:.2f} kr. ved {x_opt * 10:.0f} solgte is")
```

```python
#@title Tjek dit svar (kør cellen)
print("Rigtigt" if np.isclose(float(x_opt), 60) and np.isclose(P_max, 1120) else "Ikke endnu: tjek at du har valgt det positive kritiske punkt")
```

For at plotte skal du have talværdier. Et sympy-udtryk kan ikke regne på et helt numpy-array på én gang, så vi sætter én x-værdi ind ad gangen med en list comprehension. `dtype=float` laver resultaterne om til almindelige decimaltal. Vi bruger navnet `x_vals` til talværdierne, så vi ikke overskriver symbolet `x`.

```python
x_vals = np.linspace(0, 80, 200)
P_vals = np.array([P.subs(x, v) for v in x_vals], dtype=float)
```

```python
# Plot P_vals mod x_vals, og markér maksimum med en prik.
# Giv figuren aksetekster, overskrift og forklaring (legend).
plt.plot(x_vals, P_vals, label="Profit P(x)")
plt.plot(float(x_opt), P_max, "o", label="Maksimum")
plt.xlabel("Antal solgte is (tiere)")
plt.ylabel("Profit pr. dag (kr.)")
plt.title("Ishuset: daglig profit for en kiosk")
plt.legend()
plt.show()
```

**Svar med tekst:** Hvad sker der med profitten til venstre og til højre for maksimum, og hvordan hænger det sammen med, at $P'(x) = 0$ i punktet?

**Løsningsforslag:** Til venstre for maksimum stiger profitten, når der sælges flere is, så hældningen $P'(x)$ er positiv. Til højre falder profitten, så hældningen er negativ. I selve maksimum skifter hældningen fortegn og er derfor 0. Det er grunden til, at man finder maksimum ved at løse $P'(x) = 0$. Den negative andenafledte bekræfter, at hældningen går fra positiv til negativ, altså at punktet er et maksimum.

Blokken bygger på prikproduktet fra Del 4 i notebook 1. Du kan godt løse den, før du har lavet Del 4: opgave 4.1 forklarer det, du skal bruge.

Ishuset laver selv vaffelis, milkshakes og slushice i køkkenet i Aalborg. Matricen `A` viser, hvor meget mælk, sukker og arbejdstid én af hver vare bruger. Vektoren `b` viser, hvad køkkenet har til rådighed på en dag.

```python
# Opsætning til blok 4: kør denne celle først
import numpy as np
import sympy as sp

# Ressourceforbrug pr. vare. Kolonner: vaffelis, milkshake, slushice
A = np.array([
    [ 1,  3,  0],    # mælk (dl)
    [20, 30, 40],    # sukker (gram)
    [ 2,  3,  1],    # arbejdstid (minutter)
])
b = np.array([250, 6700, 430])     # til rådighed pr. dag: mælk, sukker, arbejdstid
priser = np.array([50, 55, 35])    # salgspris pr. vare i kr.
```

### Opgave 4.1: Ressourceforbrug med @

En produktionsplan er en vektor med antallet af hver vare. `A @ plan` ganger hver række i `A` med planen og lægger sammen, ligesom SUMPRODUKT i Excel, én gang for hver ressource. Resultatet er forbruget af mælk, sukker og arbejdstid.

Beregn forbruget for planen herunder, og find, hvor meget der er tilbage af hver ressource.

```python
plan = np.array([100, 40, 70])    # vaffelis, milkshakes, slushice

forbrug = A @ plan
rest = b - forbrug
print("Forbrug:", forbrug)
print("Tilbage:", rest)
```

```python
#@title Forventet output (klik for at vise)
print("Forbrug: [ 220 6000  390]")
print("Tilbage: [ 30 700  40]")
```

### Opgave 4.2: Find planen, der bruger det hele

Lederen vil have en plan, der bruger præcis det, køkkenet har til rådighed. Det er ligningssystemet $A\mathbf{x} = \mathbf{b}$.

1. Beregn determinanten af `A` med `np.linalg.det`.
2. Løs systemet på to måder: med `np.linalg.solve(A, b)` og med den inverse matrix, `np.linalg.inv(A) @ b`.
3. Sammenlign de to løsninger med `np.allclose`, som er `np.isclose` for hele arrays.

```python
det_A = np.linalg.det(A)
print(f"Determinant: {det_A}")

x_solve = np.linalg.solve(A, b)
x_inv = np.linalg.inv(A) @ b
print(x_solve)
print(x_inv)
print(np.allclose(x_solve, x_inv))
```

```python
#@title Tjek dit svar (kør cellen)
print("Rigtigt" if np.allclose(A @ x_solve, b) else "Ikke endnu: løsningen skal opfylde A @ x = b")
```

**Svar med tekst:** Afhængigt af computeren kan determinanten blive vist med en lille afrundingsfejl, fx 89.99999999999997 i stedet for 90. Hvorfor? Og hvad fortæller det, at determinanten ikke er 0, om systemet?

**Løsningsforslag:** `np.linalg.det` regner med decimaltal, så resultatet kan få en lille afrundingsfejl, som vi så i opgave 2.5. Determinanten er 90, altså forskellig fra 0. Det betyder, at systemet har præcis én løsning: der findes én og kun én plan, der bruger alle ressourcer fuldt ud, nemlig 100 vaffelis, 50 milkshakes og 80 slushice.

### Opgave 4.3: Kontrol og omsætning

Kontrollér, at planen bruger præcis `b`, ved at gange `A` med løsningen. Beregn derefter dagens omsætning for planen som prikproduktet af planen og priserne.

```python
# Kontrol: forbrug ved løsningen
kontrol = A @ x_solve
print(kontrol)
print(np.allclose(kontrol, b))

# Omsætning
omsaetning = x_solve @ priser
print(f"Omsætning: {omsaetning:,.0f} kr.")
```

```python
#@title Forventet output (klik for at vise)
print("[ 250. 6700.  430.]")
print("True")
print("Omsætning: 10,550 kr.")
```

### Opgave 4.4: Højere pris på milkshakes

Lederen overvejer at hæve prisen på milkshakes med 10 procent. Lav en ny prisvektor, og beregn omsætningen for den samme plan.

```python
priser_ny = priser * np.array([1, 1.1, 1])
omsaetning_ny = x_solve @ priser_ny
print(f"Ny omsætning: {omsaetning_ny:,.0f} kr.")
```

```python
#@title Forventet output (klik for at vise)
print("Ny omsætning: 10,825 kr.")
```

**Svar med tekst:** Skal du løse ligningssystemet igen, når prisen ændres? Begrund dit svar.

**Løsningsforslag:** Nej. Planen bestemmes af ressourceforbruget `A` og det, der er til rådighed, `b`. Priserne indgår ikke i ligningssystemet, så planen er den samme. Kun omsætningen ændres, fordi den beregnes ud fra planen og priserne.

### Opgave 4.5: Når systemet ikke kan løses

En kollega har lavet en ny tabel, men ved en fejl er tredje række blevet summen af de to første. Kør begge celler, og læs fejlbeskeden.

```python
B = np.array([
    [ 1,  3,  0],
    [20, 30, 40],
    [21, 33, 40],    # = række 1 + række 2
])
print(np.linalg.det(B))
```

```python
# ⚠️ Denne celle fejler med vilje
np.linalg.solve(B, b)
```

**Svar med tekst:** Hvad er determinanten, og hvad siger fejlbeskeden (nederste linje)? Forklar sammenhængen.

**Løsningsforslag:** Determinanten er 0 (eller et tal meget tæt på 0), og fejlbeskeden `LinAlgError: Singular matrix` siger, at matricen er singulær, dvs. ikke kan inverteres. Når en række er en kombination af andre rækker, indeholder den ingen ny information, og systemet har enten ingen eller uendeligt mange løsninger. Derfor kan `np.linalg.solve` ikke finde én bestemt plan.

### Opgave 4.6: Samme system med sympy

Sympy kan løse ligningssystemer eksakt med `sp.linsolve`. Den skal have totalmatricen, hvor `b` er sat på som sidste kolonne, og symbolerne for de ukendte. `np.column_stack` sætter `b` på som en ekstra kolonne.

```python
x1, x2, x3 = sp.symbols("x1 x2 x3")
totalmatrix = sp.Matrix(np.column_stack([A, b]))
totalmatrix
```

```python
loesning = sp.linsolve(totalmatrix, (x1, x2, x3))
print(loesning)
```

## Blok 5: Samlet opgave, vinterkiosken

Her er hele opgaven skrevet som tekst, og kodecellerne indeholder kun kommentarer. Find selv de mønstre, du skal bruge, i denne notebook og i notebook 1. Der er ingen tjek-celler i denne blok.

Ishuset overvejer at holde kiosken i Aalborg åben om vinteren og sælge kakao. Pr. dag regner lederen med faste omkostninger på 600 kr., variable omkostninger på 8 kr. pr. kop og en salgspris på 35 kr. pr. kop.

**Del 1: Break-even**

1. Definér $x$ som symbol og omkostningerne $C(x)$ og omsætningen $R(x)$ som udtryk.
2. Find break-even med sympy, og lav resultatet om til et decimaltal.
3. Plot $C(x)$ og $R(x)$ for 0 til 50 kopper, markér break-even, og giv figuren aksetekster, overskrift og forklaring.

**Del 2: Den bedste pris**

En markedsundersøgelse anslår, at der sælges $q(p) = 200 - 4p$ kopper pr. dag ved prisen $p$. Profitten pr. dag er derfor

$$P(p) = (p - 8)(200 - 4p) - 600$$

1. Definér $p$ som symbol og $P$ som udtryk.
2. Find den pris, der giver den største profit, og bekræft med andenafledte-testen, at det er et maksimum.
3. Beregn den største profit og antallet af solgte kopper ved den pris.
4. Plot $P(p)$ for $p$ fra 8 til 50, og markér maksimum.

**Del 3: Diskussion**

1. Den bedste pris er lavere end de 35 kr., lederen havde tænkt sig. Forklar ud fra din figur, hvorfor en lavere pris kan give en større profit.
2. Hvad sker der med den bedste pris, hvis de variable omkostninger stiger fra 8 til 10 kr. pr. kop? Prøv det i koden, og forklar.

```python
# Importér de nødvendige pakker
import numpy as np
import matplotlib.pyplot as plt
import sympy as sp
```

```python
# Del 1.1: Definér symbolet x og udtrykkene C og R
x = sp.symbols("x")
C = 600 + 8 * x
R = 35 * x
```

```python
# Del 1.2: Find break-even med sympy
be = sp.solve(sp.Eq(C, R), x)
x_be = float(be[0])
print(f"Break-even ved {x_be:.2f} kopper")
```

```python
# Del 1.3: Plot C(x) og R(x), og markér break-even
x_vals = np.linspace(0, 50, 100)
C_vals = np.array([C.subs(x, v) for v in x_vals], dtype=float)
R_vals = np.array([R.subs(x, v) for v in x_vals], dtype=float)

plt.plot(x_vals, C_vals, label="Omkostning C(x)")
plt.plot(x_vals, R_vals, label="Omsætning R(x)")
plt.plot(x_be, 35 * x_be, "o", label="Break-even")
plt.xlabel("Antal solgte kopper")
plt.ylabel("Kroner")
plt.title("Vinterkiosken: break-even")
plt.legend()
plt.show()
```

```python
# Del 2.1: Definér p som symbol og profitfunktionen P
p = sp.symbols("p")
P = (p - 8) * (200 - 4 * p) - 600
print(sp.expand(P))
```

```python
# Del 2.2: Find den bedste pris, og lav andenafledte-testen
P1 = sp.diff(P, p)
P2 = sp.diff(P1, p)
kritiske = sp.solve(P1, p)
print(kritiske)

p_opt = kritiske[0]
print(f"P''({p_opt}) = {P2.subs(p, p_opt)}")   # negativ, altså et maksimum
```

```python
# Del 2.3: Største profit og antal solgte kopper ved den bedste pris
P_max = float(P.subs(p, p_opt))
kopper = 200 - 4 * p_opt
print(f"Bedste pris: {float(p_opt):.2f} kr.")
print(f"Største profit: {P_max:.2f} kr. pr. dag")
print(f"Solgte kopper: {kopper}")
```

```python
# Del 2.4: Plot P(p), og markér maksimum
p_vals = np.linspace(8, 50, 200)
P_vals = np.array([P.subs(p, v) for v in p_vals], dtype=float)

plt.plot(p_vals, P_vals, label="Profit P(p)")
plt.plot(float(p_opt), P_max, "o", label="Maksimum")
plt.xlabel("Pris pr. kop (kr.)")
plt.ylabel("Profit pr. dag (kr.)")
plt.title("Vinterkiosken: profit som funktion af prisen")
plt.legend()
plt.show()
```

```python
# Del 3.2: Gentag beregningen af den bedste pris med variable omkostninger på 10 kr.
P_10 = (p - 10) * (200 - 4 * p) - 600
p_opt_10 = sp.solve(sp.diff(P_10, p), p)[0]
print(f"Bedste pris ved 10 kr. i variable omkostninger: {float(p_opt_10):.2f} kr.")
```

**Del 3: Svar med tekst**

**Løsningsforslag:** 1. Ved 35 kr. sælges kun 60 kopper, og profitten er 1.020 kr. Ved 29 kr. sælges 84 kopper, og profitten er 1.164 kr. Figuren viser, at profitten stiger, når prisen sænkes fra 35 kr. mod 29 kr.: de ekstra kopper, der sælges, opvejer den lavere pris pr. kop. Sænkes prisen under 29 kr., bliver fortjenesten pr. kop så lille, at profitten falder igen.

2. Den bedste pris stiger fra 29 kr. til 30 kr. Kioskens ekstra omkostning på 2 kr. pr. kop sendes altså kun halvt videre til kunderne, fordi en højere pris også betyder færre solgte kopper.

## Part 2: Lists and arrays

A **list** can hold several values. But look at what happens when you multiply a list by 2: it is **repeated**, not calculated on.

````python
# A list with three of Ishuset's prices
prices_list = [35, 45, 50]
print(prices_list * 2)
````

For calculations we use a **numpy array** instead. If you multiply an array by 2, **every number** is multiplied. That is the whole point of numpy: calculating on many numbers at once.

````python
# The same numbers as an array: now it calculates
prices = np.array([35, 45, 50])
print(prices * 2)
````

**Indexing and slicing.** Python counts from 0: `prices[0]` is the first number. A slice `prices[0:2]` gives the elements from position 0 **up to and including** position 1 (the end limit is not included). Negative indices count from the back: `prices[-1]` is the last number, `prices[-2]` the second to last.

````python
# Single elements and slices
print(prices[0])
print(prices[2])
print(prices[0:2])
print(prices[-1])    # the last number
print(prices[-2:])   # the last two numbers
````

**Element-wise calculation.** Everything you can do with one number, you can do with a whole array: multiply, add, subtract. Two arrays of the same length can also be calculated together, element by element.

````python
# All prices with a 20 percent discount
discount_prices = prices * 0.8
print(discount_prices)

# Price increase: 3 DKK on top of all prices
print(prices + 3)

# Two arrays added together, element by element
print(prices + discount_prices)
````

**Tables: 2D arrays.** An array can have rows and columns, like a small spreadsheet. Here: ice creams sold per week in the three summer kiosks. `sales[0]` is the first row, `sales[:, 0]` is the first column, `sales[1, 2]` is one particular number (row 1, column 2). Calculations also work on a single row.

````python
# Rows: Blokhus, Løkken, Skagen. Columns: week 1 to 4.
sales = np.array([
    [120, 145, 160, 150],   # Blokhus
    [ 95, 110, 125, 115],   # Løkken
    [ 80, 100, 105,  95],   # Skagen
])

print(sales[0])       # the whole Blokhus row
print(sales[:, 0])    # week 1 for all three kiosks
print(sales[1, 2])    # Løkken, week 3
print(sales[0] * 2)   # if Blokhus sold twice as much
````

## Part 3: Functions and plots

**Your own functions with `def`.** A function takes an input and gives an output, as in mathematics. Here: the daily costs of a kiosk, 400 DKK in fixed expenses plus 12 DKK per ice cream sold.

````python
# Daily costs as a function of the number of ice creams sold
def cost(x):
    return 400 + 12 * x

print(cost(0))
print(cost(10))
print(cost(100))
````

**Many x values at once: `np.linspace`.** `np.linspace(0, 40, 100)` gives 100 evenly spaced numbers from 0 to 40. If you send them through the function, you get 100 y values: everything a plot needs.

````python
# 100 x values from 0 to 40 ice creams sold
x = np.linspace(0, 40, 100)
y = cost(x)

print(x[:5])   # the first five x values
print(y[:5])   # the corresponding y values
````

**The pattern behind every function plot:** linspace, calculate, plot, text on the axes, title. It is the same five lines every time; only the content changes.

````python
# The full plot pattern
plt.plot(x, y)
plt.xlabel("Ice creams sold")
plt.ylabel("DKK")
plt.title("Ishuset: daily costs")
plt.show()
````

````python
# === TASK 1: SOLUTION ===
def revenue(x):
    return 45 * x

print(revenue(0))
print(revenue(10))
print(revenue(100))
````

````python
# === TASK 2: SOLUTION ===
x = np.linspace(0, 40, 100)
y = revenue(x)

plt.plot(x, y)
plt.xlabel("Ice creams sold")
plt.ylabel("DKK")
plt.title("Ishuset: revenue")
plt.show()
````

````python
# === TASK 3: SOLUTION ===
x = np.linspace(0, 40, 100)
plt.plot(x, cost(x), label="Cost")
plt.plot(x, revenue(x), label="Revenue")
plt.xlabel("Ice creams sold")
plt.ylabel("DKK")
plt.title("Ishuset: cost and revenue")
plt.legend()
plt.show()
````

````python
# === TASK 4: SOLUTION ===
x_be = 400 / 33
print(f"Break-even at {x_be:.1f} ice creams sold")

x = np.linspace(0, 40, 100)
plt.plot(x, cost(x), label="Cost")
plt.plot(x, revenue(x), label="Revenue")
plt.plot(x_be, revenue(x_be), "o")
plt.xlabel("Ice creams sold")
plt.ylabel("DKK")
plt.title("Ishuset: break-even")
plt.legend()
plt.show()
````

**Answer in words (suggested solution):** An example of an answer: The kiosk has to sell about 12.1 ice creams a day (13 in practice) before the revenue covers the costs; if it sells fewer, the day runs at a loss. Other wordings with the same meaning are just as good.

**(extra) Task 5 (suggested solution):** Break-even moves to the right, to 600/33, about **18.2 ice creams**. That makes sense: higher fixed expenses require more ice creams sold before the day breaks even. The cost curve starts higher but has the same slope.

## Part 4: Dot product and outlook

**Dot product: Excel's SUMPRODUCT in Python.** A family orders 2 units of Waffle 2 scoops, 1 softice and 1 slushice. The bill is quantity times price, added up. First in two steps, then with the shortcut `@`, which does both at once.

````python
# The family's order
quantity = np.array([2, 1, 1])
prices   = np.array([50, 45, 35])

# Step by step: element-wise product, then sum
print(quantity * prices)
print((quantity * prices).sum())

# The shortcut: dot product with @
print(quantity @ prices)
````

**For demonstration only: tables as matrices.** In Applied Mathematics in the 1st semester, 2D arrays are called **matrices**, and there you calculate with them systematically. Here you see the shape of the table and a swap of two rows. You do not need to be able to write this yourself, only recognize it.

````python
# FOR DEMONSTRATION ONLY
print(sales.shape)      # (3, 4): 3 rows, 4 columns
print(sales[[1, 0, 2]]) # the rows swapped: Løkken at the top
````

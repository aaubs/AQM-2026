## Toolbox

The functions you need for Workshop 3 (hypothesis tests). The lecture notebook shows them in use.

| Function | What it does |
|---|---|
| `np.array([...])` | Turn a list of numbers into a sample |
| `len(x)`, `x.mean()` | Sample size \(n\) and sample mean \(\bar{x}\) |
| `x.std(ddof=1)`, `x.var(ddof=1)` | Sample SD \(s\) and variance \(s^2\) (divide by \(n-1\)) |
| `np.sqrt(n)` | Square root (for standard errors) |
| `norm.ppf(1 - alpha)` | Critical \(z_\alpha\) (upper-tailed \(z\)-test) |
| `norm.ppf(alpha)` | Critical \(-z_\alpha\) (lower-tailed \(z\)-test) |
| `norm.ppf(1 - alpha/2)` | Critical \(z_{\alpha/2}\) (two-sided \(z\)-test) |
| `norm.cdf(z)` | Lower-tailed \(p\)-value for a \(z\)-test: \(P(Z \le z)\) |
| `norm.sf(z)` | Upper-tailed \(p\)-value: \(P(Z > z)\) (same as `1 - norm.cdf(z)`) |
| `stats.t.ppf(1 - alpha, df=n-1)` | Critical \(t_{n-1,\alpha}\) (upper-tailed \(t\)-test) |
| `stats.t.ppf(alpha, df=n-1)` | Critical \(-t_{n-1,\alpha}\) (lower-tailed \(t\)-test) |
| `stats.t.ppf(1 - alpha/2, df=n-1)` | Critical \(t_{n-1,\alpha/2}\) (two-sided \(t\)-test) |
| `stats.t.cdf(t, df=n-1)` | Lower-tailed \(p\)-value for a \(t\)-test |
| `stats.t.sf(t, df=n-1)` | Upper-tailed \(p\)-value for a \(t\)-test |
| `stats.chi2.ppf(1 - alpha, df=n-1)` | Critical \(\chi^2_{n-1,\alpha}\) (upper-tailed variance test) |
| `stats.chi2.sf(chi2, df=n-1)` | Upper-tailed \(p\)-value for a variance test |



### mean of a normal distribution 

the population variance is known

$$
z = \frac{\bar{x} - \mu_0}{\sigma / \sqrt{n}}
$$

the population variance is unknown

- In a \(t\)-test, **\(s\)** is the **sample standard deviation** 
When $\sigma$ is unknown we use the sample standard deviation $s$ in its place. The test statistic is

$$
t = \frac{\bar{x} - \mu_0}{s/\sqrt{n}}, \qquad \text{df} = n - 1,
$$


### Large-sample $z$-tests for a population proportion $P$

Let $\hat p = x/n$ be the sample proportion and $P_0$ the hypothesised value. Under $H_0$, when $nP_0(1-P_0)$ is large enough (rule of thumb: $nP_0 \ge 5$ and $n(1-P_0) \ge 5$; stricter books use 10 or $nP_0(1-P_0) > 5$), the statistic

$$
z = \frac{\hat p - P_0}{\sqrt{P_0(1-P_0)/n}}
$$


### $\chi^2$-test for the variance of a normal population

For a random sample of size $n$ from a **normal** population with sample variance $s^2$, the statistic

$$
\chi^2 = \frac{(n-1)s^2}{\sigma_0^2}
$$

# Saving and loading a cleaned dataset in Google Drive

[![Open In Colab](https://colab.research.google.com/assets/colab-badge.svg)](https://colab.research.google.com/github/aaubs/AQM-2026/blob/main/lecture10/save_and_load_drive.ipynb)

Applied Quantitative Methods, Lecture 10. A short demonstration for Colab.

We recommend one notebook that runs from top to bottom: from the raw data, through the cleaning, to the last
analysis. Then nothing needs to be saved. When you split the work, for example one notebook for cleaning and one for
the analysis, you need to save the cleaned data somewhere the next notebook can find it. In Colab, the place to use is
your Google Drive.

The example uses the messy order export from Lecture 5 and one cleaning step.

## 1. Load the raw data and clean it

```python
import os
import pandas as pd

url = "https://raw.githubusercontent.com/aaubs/AQM-2026/main/data/"
orders = pd.read_csv(url + "orders_export.csv")
print(f"Raw data: {orders.shape[0]} rows")

clean = orders.drop_duplicates()          # one cleaning step, as an example
print(f"Cleaned data: {clean.shape[0]} rows")
```

## 2. Connect Colab to your Google Drive

Run the cell. Colab asks for permission to use your Google Drive; accept it with the account you use for Colab. Your
Drive then appears as a folder called `/content/drive/MyDrive`.

```python
from google.colab import drive
drive.mount("/content/drive")
```

## 3. Save the cleaned data in your Drive

The cell makes a folder `AQM` in your Drive (if it is not there already) and saves the data as a CSV file in it.
`index=False` leaves out the row numbers, which would otherwise come back as an extra column.

```python
folder = "/content/drive/MyDrive/AQM"
os.makedirs(folder, exist_ok=True)

path = folder + "/orders_clean.csv"
clean.to_csv(path, index=False)
print("Saved:", path)
print(os.listdir(folder))
```

## 4. Read it back

This is what the next notebook does, or this notebook in a later session: connect to Drive (step 2), then read the
file. You can also find the file in Google Drive in your browser, in the folder `AQM`.

```python
clean_again = pd.read_csv("/content/drive/MyDrive/AQM/orders_clean.csv")
print(f"Read back: {clean_again.shape[0]} rows")
clean_again.head()
```

## Without Drive

`clean.to_csv("orders_clean.csv", index=False)` without a Drive folder saves the file in the session's temporary
storage, the folder icon on the left in Colab. It is deleted when the session ends. To keep it, download it:

```python
from google.colab import files
clean.to_csv("orders_clean.csv", index=False)
files.download("orders_clean.csv")
```

## Code patterns

Two complete patterns for today. Replace `df`, `y`, `x1` and `x2` with your data and column names. The patterns are
templates; they do not run as they stand.

**Fit, check, and refit with robust standard errors**

```python
model = smf.ols("y ~ x1 + x2", data=df).fit()

plt.scatter(model.fittedvalues, model.resid, s=8, alpha=0.5)    # residual plot
plt.axhline(0, color="darkorange", linestyle="--")
plt.xlabel("Fitted value"); plt.ylabel("Residual")
plt.show()

lm, lm_p, f, f_p = het_breuschpagan(model.resid, model.model.exog)
print(f"Breusch-Pagan = {lm:.2f}, p = {lm_p:.2g}")
jb, jb_p = stats.jarque_bera(model.resid)
print(f"Skew = {stats.skew(model.resid):.2f}, Kurtosis = {stats.kurtosis(model.resid, fisher=False):.2f}, "
      f"Prob(JB) = {jb_p:.2f}")

robust = smf.ols("y ~ x1 + x2", data=df).fit(cov_type="HC1")
print(pd.DataFrame({"coef": model.params, "se classical": model.bse, "se robust": robust.bse}))
```

**A log model, a prediction, and back to the original unit**

```python
df["log_y"] = np.log(df["y"])
model = smf.ols("log_y ~ x1 + x2", data=df).fit(cov_type="HC1")
print(model.summary())
b = model.params["x1"]
print(f"about {100 * b:.1f}%, exactly {100 * (np.exp(b) - 1):.1f}% per unit of x1")

new = pd.DataFrame({"x1": [5], "x2": [40]})      # one column for every x in the formula
print(np.exp(model.predict(new)))                # predicted y in the original unit
```

## Setup

````python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy import stats
import statsmodels.formula.api as smf
from statsmodels.stats.diagnostic import het_breuschpagan
from statsmodels.stats.outliers_influence import variance_inflation_factor

url = "https://raw.githubusercontent.com/aaubs/AQM-2026/main/data/"
customers = pd.read_csv(url + "customers_sample.csv")   # 1,000 RetailCorp customers drawn at random
customers["log_spend"] = np.log(customers["spend_2025"])
customers["log_visits"] = np.log(customers["visits_2025"])
customers.head()
````

### Task 1. Tenure among customers of the same age (worked)

The lecture's multiple regression in code. Run the cells and answer the questions below them.

y = `log_spend` (ln of spend in DKK in 2025); x<sub>1</sub> = `tenure_years` (years as a customer); x<sub>2</sub> =
`age` (years).

````python
y_col = "log_spend"      # step 3: change to "spend_2025"

simple = smf.ols(f"{y_col} ~ tenure_years", data=customers).fit()
multiple = smf.ols(f"{y_col} ~ tenure_years + age", data=customers).fit()
print(multiple.summary())
````

````python
b_simple = simple.params["tenure_years"]
b1, b2 = multiple.params["tenure_years"], multiple.params["age"]
d = smf.ols("age ~ tenure_years", data=customers).fit().params["tenure_years"]

print(f"Tenure alone: {b_simple:.4f}   with age: {b1:.4f}   age: {b2:.4f}")
print(f"b1 + b2 * d = {b1:.4f} + {b2:.4f} * {d:.4f} = {b1 + b2 * d:.4f}")
print(f"The tenure coefficient falls to {b1 / b_simple:.0%} of its simple value")
if y_col == "log_spend":
    print(f"Tenure with age, as a percent: about {100 * b1:.1f}%, exactly {100 * (np.exp(b1) - 1):.1f}% per year")
````

<details>
<summary>Hint 2</summary>

The first code pattern does questions 1, 2 and 4. For question 3: `customers["resid"] = model.resid`, then

````python
levels = smf.ols("spend_2025 ~ tenure_years + age", data=customers).fit()

plt.scatter(levels.fittedvalues, levels.resid, s=8, alpha=0.5)
plt.axhline(0, color="darkorange", linestyle="--")
plt.xlabel("Fitted value, DKK"); plt.ylabel("Residual, DKK")
plt.show()

lm, lm_p, f, f_p = het_breuschpagan(levels.resid, levels.model.exog)
jb, jb_p = stats.jarque_bera(levels.resid)
print(f"Breusch-Pagan = {lm:.2f}, p = {lm_p:.1e}")
print(f"Skew = {stats.skew(levels.resid):.2f}, Kurtosis = {stats.kurtosis(levels.resid, fisher=False):.2f}, "
      f"Prob(JB) = {jb_p:.2f}")
````

````python
# residuals by region
customers["resid"] = levels.resid
customers.boxplot(column="resid", by="region", figsize=(8, 4))
plt.ylabel("Residual, DKK")
plt.show()
print(customers.groupby("region")["resid"].std().round(0))
````

````python
# robust standard errors
levels_r = smf.ols("spend_2025 ~ tenure_years + age", data=customers).fit(cov_type="HC1")
print(pd.DataFrame({"coef": levels.params, "se classical": levels.bse, "se robust": levels_r.bse}).round(3))
print(levels_r.conf_int().round(2))
````

````python
# the same model in logs
logs = smf.ols("log_spend ~ tenure_years + age", data=customers).fit()
lm, lm_p, f, f_p = het_breuschpagan(logs.resid, logs.model.exog)
jb, jb_p = stats.jarque_bera(logs.resid)
print(f"Logs: Breusch-Pagan = {lm:.2f}, p = {lm_p:.2f}")
print(f"Skew = {stats.skew(logs.resid):.2f}, Kurtosis = {stats.kurtosis(logs.resid, fisher=False):.2f}, "
      f"Prob(JB) = {jb_p:.2f}")
````

1. No curve, but a clear fan: the residuals spread out as the fitted spend grows.
2. Breusch-Pagan = 21.41 (2 degrees of freedom), p = 2.2e-05: we reject equal scatter. Skew 2.02 is well beyond 1,
   so the residuals are strongly skewed to the right; Kurtosis 10.56 is far above 3, so a few customers lie far
   above the line; and Prob(JB) = 0.00, so the test rejects normal errors. The straight line holds; equal scatter and normal errors
   fail. Independence comes from the random sample.
3. The box plots look much alike. The standard deviation of the residuals is between about 1,240 and 1,520 DKK in
   all five regions, and every region has a few customers far above the line. The scatter grows with the fitted
   value and hardly differs between regions.
4. The coefficients do not change. The robust standard error of tenure is 22.28 against 18.23 (classical), so t
   falls from 7.66 to 6.27 and the interval widens from 103.9 to 175.4 to about 96.0 to 183.3 DKK per year. The
   conclusion stands: among customers of the same age, longer-standing customers spend more, about 140 DKK per year
   of tenure.
5. In logs, Breusch-Pagan = 2.69, p = 0.26: the test does not reject equal scatter. Skew 0.02, Kurtosis 3.08 and
   Prob(JB) = 0.84: the residuals are close to normal. Taking logs removed the fan and the skew.

### Task 4. Predictions for two customers

Use the final model from the lecture: ln spend on `tenure_years`, `age` and `visits_2025`. Predict the spend in DKK of
two customers:

<details>
<summary>Hint 1</summary>

The model predicts ln spend; the exponential takes it back to DKK. A difference in ln is a ratio in the original
unit: exp(ln y<sub>B</sub> − ln y<sub>A</sub>) = y<sub>B</sub> / y<sub>A</sub>.
</details>

<details>
<summary>Hint 2</summary>

`pd.DataFrame({"tenure_years": [1, 8], "age": [25, 60], "visits_2025": [30, 20]})`, `model.predict(...)`,
`np.exp(...)`, `customers["spend_2025"].mean()` and `.median()`, and `((cond1) & (cond2)).sum()` to count rows.
</details>

````python
final = smf.ols("log_spend ~ tenure_years + age + visits_2025", data=customers).fit()
new = pd.DataFrame({"tenure_years": [1, 8], "age": [25, 60], "visits_2025": [30, 20]}, index=["A", "B"])
pred_log = final.predict(new)
print(pd.DataFrame({"ln spend": pred_log.round(4), "spend, DKK": np.exp(pred_log).round(0)}))

print(f"B / A = {np.exp(pred_log['B']) / np.exp(pred_log['A']):.2f}, "
      f"exp(difference) = {np.exp(pred_log['B'] - pred_log['A']):.2f}")
print(f"Sample mean {customers['spend_2025'].mean():,.0f} DKK, median {customers['spend_2025'].median():,.0f} DKK")

like_a = ((customers["tenure_years"] <= 2) & (customers["age"] <= 30)).sum()
like_b = ((customers["tenure_years"] >= 6) & (customers["age"] >= 55)).sum()
print("Customers like A:", like_a, " like B:", like_b)
````

1. A: ln spend 6.985, about 1,080 DKK. B: ln spend 7.789, about 2,413 DKK.
2. B is predicted to spend 2.23 times as much as A. The difference in predicted ln spend is 0.804, and
   exp(0.804) = 2.23: in a log model, differences in ln are ratios in DKK.
3. A is predicted well below both the sample mean (2,162 DKK) and the median (1,794 DKK); B above both. Spend is
   skewed: among customers like B, a few spend much more than the rest and pull the average up. The model predicts the
   average ln spend, and the exponential of an average of logs lies near the middle of the customers (a median), below
   their average spend. So 2,413 DKK is a typical spend for customers like B, and their average is somewhat higher. If the errors in logs
   are normal, the average is about exp(s<sub>e</sub>²/2) times the prediction, with s<sub>e</sub> = 0.54 here:
   about 1.16 × 2,413 ≈ 2,790 DKK.
4. 102 customers have at most 2 years and are at most 30, and 82 have at least 6 years and are at least 55. Both
   profiles are well inside the data, so these are not extrapolations.

## Extension tasks

### Extension 1. Age when they became customers

Make a new column with each customer's age at the first order, `age_first = age - tenure_years`, and add it to the
model for ln spend with `tenure_years` and `age`. Look at the coefficients and at the notes under the output. What
happened, and why can the data not separate the three variables?

````python
customers["age_first"] = customers["age"] - customers["tenure_years"]
m_perfect = smf.ols("log_spend ~ tenure_years + age + age_first", data=customers).fit()
print(m_perfect.summary())
````

age_first is exactly age minus tenure, so one of the three variables is always determined by the other two: perfect
multicollinearity. The data cannot say how much of a difference belongs to each, because no two customers differ in
one of them while the other two stay the same. statsmodels still prints coefficients, but they are one arbitrary
split among many that fit equally well. The standard errors and p-values look ordinary but mean nothing here. The
warning and note [2] under the output say that the design matrix is singular (the smallest eigenvalue is close to 0). The fix is to leave one of the three out; with tenure and age in the model,
age at the first order adds no information.

### Extension 2. The VIF by hand

For the model of ln spend on `tenure_years` and `age`, regress `age` on `tenure_years`, take its R², and compute
VIF = 1 / (1 − R²). Check the result with `variance_inflation_factor`. Then do the same for the model that also has
`visits_2025`: regress each x on the other two.

````python
r2_age = smf.ols("age ~ tenure_years", data=customers).fit().rsquared
print(f"Two x: R2 = {r2_age:.4f}, VIF = {1 / (1 - r2_age):.3f}")

three = smf.ols("log_spend ~ tenure_years + age + visits_2025", data=customers).fit()
X = three.model.exog
for i, name in enumerate(["tenure_years", "age", "visits_2025"], start=1):
    others = " + ".join(n for n in ["tenure_years", "age", "visits_2025"] if n != name)
    r2 = smf.ols(f"{name} ~ {others}", data=customers).fit().rsquared
    print(f"{name}: by hand {1 / (1 - r2):.3f}, variance_inflation_factor {variance_inflation_factor(X, i):.3f}")
````

With two x's, R² of age on tenure is 0.2354 and VIF = 1 / (1 − 0.2354) = 1.308, the same as `variance_inflation_factor`.
With three x's the VIFs are 1.34 (tenure), 1.41 (age) and 1.08 (visits). All are far below 5: the three variables
are related, but the data can separate them.
